## Exercise 2 - Determinants

Two vectors $(a, b)$ and $(c, d)$ form the matrix $\begin{bmatrix}a & b\\c & d &\end{bmatrix}$. Its determinant is $D = ad - bc$. 

Given a pair of simultaneous equations:

$$ ax + by = e \\ c x + dy = f $$

We can find solutions using Cramer's rule: 

$$ x = (ed-bf) / D \\ y = (af-ec) / D $$


Note however, this only works if the determinant is not zero -- in that case, we get a division by zero error. The code below has functions for calculating the determinant and solving the equations. However, it doesn't have any check for when the determinant is zero, so we get errors on some of our tests. Can you edit the code below to include the necessary checks?

````python
def determinant(a, b, c, d):
    return a*d - b*c

def solve(a, b, c, d, e, f):

    x = (e*d - b*f) / determinant(a, b, c, d)
    y = (a*f - e*c) / determinant(a, b, c, d)

    return (x, y)

````
