## `if`

Often in our code, we'll want certain instructions (or sequence of instructions) to execute conditionally-for example, we should only run the code for processing input if we've first checked whether we've actually got some input to process. We can do this in Python using the *if-then-else* syntax. 

In general:

```python
if (test expression):
    # something that only gets done if the test expression returns True
    # another thing that only happens if it's True
# this expression happens anyway, True or False
```
The block of code identified by being _indented_ is the bit that gets switched on or off by the `if`.  This is a bit like the body of a function, where indentation was used to identify which bits of code are part of the function and why aren't.  Let's try an example.

**Comprehension check** Play with the numbers in the code below and see what happens.

````python
a = 5
b = 3
print('a is',a)
print('b is',b)
if a>=b:
    print('a is greater than or equal to b')
if a<=b:
    print('a is less than or equal to b')
if a==b:
    print('a equals b')
    print('a and b have the same value')
print('Finished')
````

## `if` `elif` and `else`

`elif` is short for "else if"

```Python
if test1:
    # this runs if test 1 passes, _i.e._ test1 is True
elif test2:
    # this runs if test 1 fails and then test 2 passes
else:
    # this runs if test 1 fails and then test 2 fails
```

You can have arbitrary numbers of `elif` statements after an `if`, and the final `else` is optional. Note that if Python finds a true condition (and executes the associated code block), it won't even bother to check the remaining tests.

Let's see some examples.

**Comprehension Check** Change the variable settings and see what happens

````python
detected_item = 'Mouse' # try also 'Dog' and 'Human'

if detected_item=='Human':
    print('Human detected: robot safety protocol engaged')
elif detected_item=='Dog':
    print('Dog detected: pet engagement protocol engaged')

print('Carrying on')
````

Next is an example with loads of `elif` statements to capture multiple cases, plus a final `else` to catch anything unexpected, which is good practice.

This is an example of a _finite state machine_, which is a handy framework for stitching together different behaviours for some sort of artificial agent.

````python
mode = 'drive' # try also 'explore', 'drive', 'goal', 'lost', and 'rubbish', or indeed anything else

if mode=='avoid':
    print('Backing up robot')
    mode = 'explore'
elif mode=='explore':
    print('Random turn')
    mode = 'drive'
elif mode=='drive':
    print('Going forwards')
    mode = 'goal'
elif mode=='goal':
    print('Turning to goal')
    mode = 'drive'
elif mode=='lost':
    print('Stop and reset')
    mode = 'explore'
else:
    print('Unexpected mode:', mode)
    raise ValueError('Unexpected mode value')
````

Now another common situation - saturation or clipping.  Suppose we want to ensure that a value is always within a certain range.  If it goes too high, we limit to the upper limit, and if too low, we limit to the lower limit.  This is really common in control where actuators typically have a fixed range of operation, _e.g._ voltage outputs or motor speeds.

**Comprehension Check** Can you make the code below a little more efficient?  How and why?

````python
command = -1.2 # test with values from -2.0 to 2.0

def saturate(command):
    print('Before saturation command is', command)

    if command > 1.0:
        command = 1.0
    if command < -1.0:
        command = -1.0

    print('After saturation command is', command)

    return command

command = saturate(command)
print(command)
````

## Nesting

Remember that any statements in the indented bit after an `if` only runs if the `if` expression is `True`.  Now we will put more `if` statements in that indented area.

```Python
if test1:
    if test2:
        # this runs if test1 is True and test2 is True
    else:
        # this runs if test1 is True and test2 is False

    # this runs if test1 is false, without even calculating test2
```

## If and Functions

We can of course use 'If' statements within function bodies.

````python
def is_positive(number):
    if number > 0:
        return True
    else:
        return False
````

One thing to be mindful of is that execution of a *return* statement will cease execution of the function. This means any code beneath the the return statement will not be run. So the example above, could be written as:

````python
def is_positive(number):
    if number > 0:
        return True
    return False #This line only runs if the line above hasn't already ended the function call.
````

### Testing Functions with 'If' statements

Now that we've seen how to use both functions and 'If' statements, we can use a simple approach to begin testing whether a function is producing the correct output.


````python
def addition(a, b):
    return a+b

if addition(1, 1) != 2:
    print("Error: 1+1 should equal 2")

if addition(1, 2) != 3:
    print("Error: 1+2 should equal 3")  

if addition(0, 1) != 1:
    print("Error: 0 +1 should equal 1")
````

Here, we've used If statements to check whether the output matches what we expect for some simple cases (i.e where we know the answer). The benefit of testing in this way is that if we change the code, we can then quickly re-run the tests and make sure our changes haven't broken things that were previously working. 

We can also put our tests into a function to make the code a bit neater.


````python
def addition(a, b):
    return a+b

def test_addition():
    
    if addition(1, 1) != 2:
        print("Error: 1+1 should equal 2")
        return False

    if addition(1, 2) != 3:
        print("Error: 1+2 should equal 3")
        return False

    if addition(0, 1) != 1:
        print("Error: 0 +1 should equal 1")
        return False
    
    return True

if test_addition():
    print("All tests passed")
````

## Summary

- Python's `if` statement can be used to run a block of code only if a condition is met
- The block is marked by indenting.  Anything indented after the `if` will only run if it passes the test, otherwise it skips to the next line at the same indentation level as the 'if'.

````python
a=1
if a==1:
    print('a is 1')
print('This gets printed regardless of the value of a')
````

- Following an `if` with an `else` means the indented bits after the `else` only run if the first test fails

````python
a=2
if a==1:
    print('a is 1')
else:
    print('a is not 1')
print('This gets printed regardless of the value of a')
````

- Following an `if` with an `elif` is short for "else if" and will run a block of code only if the first test fails but the second one passes

````python
a=0.8
if a==1:
    print('a is 1')
elif a>0.5:
    print('a is not 1')
    print('but it is bigger than a half')
````

- You can nest one if statement in another if statement

````python
a=0.9
if a>0.5:
    print('a is bigger than a half')
    if a>0.75:
        print('Not only that, but it is bigger than three quarters')
````
