# How a Computer Represents Data

Computers store data by representing values as binary numbers. A bit (short for 'binary digit') is the smallest unit of data that a computer can store. A bit can have one of two possible values: 0 or 1. We can't store much with just a single bit, but sequences of bits can be used to store any kind of information you want.

## Binary

In daily life we mostly use the **decimal** or **base-10** number system. Each number is formed by finding the sum of base-10 numbers, each multiplied by a factor (0-9).

Consider the number 104.02

|  **Unit**               |  $10^2$ | $10^1$ | $10^0$  | $10^{-1}$ | $10^{-2}$ | 
| :----------     | :------ | :----- | :--     | :-----  | :------ |
| **Value**| 100     |10      | 1       | 0.1     | 0.01    |    
| **Factor**      | 1       | 0      | 4       | 0       | 2       |
| **Total**       | 100     | 0      | 4       | 0       | 0.02    |

Sum of columns = 100 + 4 + 0.02 = 104.02


Numbers are stored on a computer as **binary** or **base-2** numbers. Each number is formed by finding the sum of base-2 numbers, each multiplied by a factor **1 or 0**. The factors are like "on" (1) and "off" (0) switches for the different columns

Consider the number 25

| **Unit**         |  $2^4$  | $2^3$  | $2^2$   | $2^1$   | $2^0$   | 
| :----------     | :------ | :----- | :------ | :-----  | :------ |
| **Value**| 16      |8       | 4       | 2       | 1       |    
| **Factor**      | 1       | 1      | 0       | 0       | 1       |
| **Total**       | 16      | 8      | 0       | 0       | 1       |

Binary representation: 11011

Sum of columns = 16 + 8 + 1 = 25

Any whole number can be represented by 1s and 0s providing we have enough columns (binary digits).

## Integers

An *Integer* data type uses a fixed number of bits to store a a whole number. The size of the number than we can store depends on how many bits we've got. For example, if we've got 4 bits, then the highest number we can store is `1111` = $1 + 2 + 4 + 8 = 15$. In general, if we have $N$ bits, then we can store $2^N$ different values, with a maximum value of $2^N-1$.

| # of bits | Min Value | Max value |
| -- | -- | -- |
| 1 | 0 | 1 |
| 2 | 0 | 3 |
| 4 | 0 | 15 |
| 8 | 0 | 255 |
| 16 | 0 | 65535 |
| 64 | 0 | 18446744073709551615 |

### Representing Negative Numbers

Integers let us represent positive numbers. To represent a negative number, we need to use one bit to store the number's sign. Using a sign bit reduces the number of bits we have available for numbers, so the maximum value is reduced to $2^{N-1} - 1$. Typically the left-most bit (also known as the most significant bit) is used as a sign bit and represents the value *-128*. So if the left-most bit is set, we start at a value of -128 and then add from there. If it's not set, we start at zero instead. 

For example, the number 11001100 = -128 + 64 + 8 + 4 = -52 and 01001100 = 64 + 8 + 4 = 76.

### How Python handles Ints.

In many programming languages, we have to decide how many bits we want to use for an Integer, and whether it should be signed or unsigned when we first create the variable. Common choices are 8-bit integers, 16 bit integers, or 32 bit integers. 

| # of bits | signed or unsigned? | Minimum Value | Maximum Value | Number of values represented | Common Name |
| -- | -- | -- | -- | -- | -- |
| 8  | unsigned | 0 | 255 | 256 | uint8 |
| 8  | signed   | -128 | 127 | 256 | int8 |
| 16 | unsigned | 0 | 65535 | 65536 | uint16 |
| 16 | signed   | -32768 | 32767 | 65536 | int16 |
| 32 | unsigned | 0 | 4294967295 | 4294967296 | uint32 |
| 32 | signed   | -2147483648 | 2147483647 | 4294967296 | int32 |
| 64 | unsigned | 0 | 18446744073709551615 | 18446744073709551616 | uint64 |
| 64 | signed   | -9223372036854775808 | 9223372036854775807 | 18446744073709551616 | int64 |

Attempting to assign a number larger than the number of bits we have would lead to a pernicious error known as Integer overflow. For example, if we add one to an unsigned 8 bit integer with a value of 255, we'll get 0 as the answer.

$uint8 (255) + uint8 (1) = 11111111 + 00000001 = 00000000 = 0$

For signed integers, the overflow takes us all the way to the minimum value.

$int8(127) + int8(1) = 01111111 + 00000001 = 10000000 = -128$

Fortunately, Python's Integer class works slightly differently, growing to represent numbers as large as we need (the only real limit is the amount of memory available.), so we don't need to worry about overflows when we use Python's integer data type. Later on in the course, we'll introduce Numpy which uses fixed-width integers like we've been describing here -- and these can overflow. 

## Floats

Real numbers are typically stored as floating point values.  You're probably already familiar with scientific notation:

$
\boxed{\underbrace{M}_{mantissa} \times \underbrace{10}_{base}\overbrace{^E}^{exponent}}
$

- $ 1.234 \times 10^0 = 1.234 \\$
- $ 2.34 \times 10^1 = 23.4 \\$
- $ 5.987 \times 10^2 = 598.7 $

To be compatible with digital "on" or "off" memory, computers use a binary form of the same:

$
\boxed{\underbrace{M}_{mantissa} \times \underbrace{2}_{base}\overbrace{^E}^{exponent}}
$


- $
1.25 = 1\frac{1}{4} \times 2^0 
\textrm{ written as 4-bit mantissa }    \boxed{ \underbrace{1}_{1}.\underbrace{0}_{\frac{1}{2}}\underbrace{1}_{\frac{1}{4}}\underbrace{0}_{\frac{1}{8}} } 
\textrm{ and 2-bit exponent }   \boxed{ \underbrace{0}_{2}\underbrace{0}_{1} }
$
- $
5.5 = 1\frac{3}{8} \times 2^2 
\textrm{ written as 4-bit mantissa }    \boxed{ \underbrace{1}_{1}.\underbrace{0}_{\frac{1}{2}}\underbrace{1}_{\frac{1}{4}}\underbrace{1}_{\frac{1}{8}} } 
\textrm{ and 2-bit exponent }   \boxed{ \underbrace{1}_{2}\underbrace{0}_{1} }
$
- $
12 = 1\frac{1}{2} \times 2^3 
\textrm{ written as 4-bit mantissa }    \boxed{ \underbrace{1}_{1}.\underbrace{1}_{\frac{1}{2}}\underbrace{0}_{\frac{1}{4}}\underbrace{0}_{\frac{1}{8}} } 
\textrm{ and 2-bit exponent }   \boxed{ \underbrace{1}_{2}\underbrace{1}_{1} }
$

This representation allows us to store much larger numbers than with Integers. A 32 bit float can store numbers upto ~ $3.4 \times 10^{38}$, while a 64 bit float (sometimes known as a double) can store numbers upto ~ $1.8 \times 10^{308}$. Python uses 64 bit floats by default (1 sign bit, 11 exponent bits, 52 mantissa bits).



### Floating Point Precision

The increased range of a floating point representation comes at the cost of *precision*. A float can represent values upto $1.8 \times 10^{308}$, but importantly, it can't represent *all* values between 0 and $1.8 \times 10^{308}$. Let's look at an example. With our 4-bit mantissa and 4-bit exponent, the bits 1010 mean 1.010 = 1.25, so with an exponent of 0110 (= 6) the value is 1.25 × 2⁶ = 80. Add one to the mantissa and 1011 = 1.375, giving 1.375 × 2⁶ = 88. There's a gap of 8 between them that we simply cannot represent — no value between 80 and 88 exists in this format. Now run the same one-bit mantissa step with a larger exponent, 1010 (= 10): 1.25 × 2¹⁰ = 1280 and 1.375 × 2¹⁰ = 1408 — the same smallest change is now a gap of 128. The exponent scales the mantissa, so a larger exponent reaches larger numbers but spreads the representable values further apart (lower precision). Adding more mantissa bits shrinks the gap but never closes it.

The limited precision of floating point numbers is why we see unexpected (but technically correct) behaviours such as:

````python
print (0.1 + 0.2 == 0.3)
print (sum([0.1]*10) == 1.0)
print (1e308 + 1e9 == 1e308) #At this scale, the gap between numbers is over a billion!
print (1.000000001 - 1.0) #This should equal 1e-9 -- what will happen to errors like this when we're making millions of these kinds of calcutions per second?
print((0.1 + 0.2) + 0.3 == 0.1 + (0.2 + 0.3)) # Addition isn't associative for floats!
````

### Inf and NaN (Not a Number)

In other programming languages, when we exceed the storage capacity of a numerical data type, we can overflow -- the number wraps back around to it's minimum value. Python works a bit differently for floats. If we exceed the maximum value (1e308), we'll get the result 'inf' - essentially Python's version of infinity. 

On the other hand, if we try to perform an operation which isn't well defined (like adding or subtracting two infinities), we'll get 'NaN' -- which stands for "Not a Number". Performing further operations on 'NaN' will still give us 'NaN'. Thus a common issue in numerical Python programming is trying to figure out which of a sequence of operations caused the first 'NaN' -- as after that, it will just propagate through all results.

````python
print(1e308 * 10)
print(float('inf') - float('inf')) #This is NaN - 'Not a number' -- addition and subtraction are not well defined for infinite quantities!
print(float('nan') == float('nan')) #NaN isn't equal to anything -- not even itself.


print(float('nan') + 5)
````

## Characters

Characters (single letters) in Python (and most other programming languages) are just integers with a look-up table associated with them. We can use the built-in Python function ```ord``` to convert a character into its numerical representation.

````python
print(ord('a'))
print(ord('A'))

print(ord('a') + ord('b')) #These are just numbers, so we can add them together if we want.
````

Similarly, we can use the built-in function ```chr``` to convert a number into a character. 

````python
print(chr(97))
print(chr(65))

print(chr(195))
````

### ASCII

There are many different tables mapping we could use to map numbers to letters. The table that was originally used for this was called 'ASCII', which is 7 bits (so 128 different characters could be represented). Clearly, this is insufficient to represent all the possible letters globally -- so ASCII has been extended with a much bigger table (Unicode) and representation scheme (UTF-8). These extended tables are compatible with ASCII though (i.e they agree that the letter 'a' is 97 and so on). The details of Unicode and UTF-8 are beyond this course, but you shouldn't be put off by these terms if you see them.

## Wrap-up

We've seen today that computers represent data as binary sequences. The same sequence can in fact represent multiple different thing -- e.g. 01000001 could represent the integer 65, the character 'A' or the floating point number 1 (assumning 4 bit mantissa, 0100 = 0.5 and 4 bit exponent 0001 = 1). This is why we need to keep track of the *data type* of every variable stored in memory -- if we don't know the data type, we don't know what the sequence of bits means!
