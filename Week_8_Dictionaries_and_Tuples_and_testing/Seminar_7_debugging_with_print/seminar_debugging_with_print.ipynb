## Debugging

Real coding is usually a multi-step process:
1. Add some functionality to your code
2. Test it
3. If it doesn't work, attempt to fix it, and repeat Step 2.
4. If it does work, go to Step 1.

A script or function will grow in little steps, each one concluding only when it is known to be correct.

The process of Step 3 is known as _debugging_.

| **Type**    | **Example** |

| Runtime errors | Accessing a list index that doesn't exist     |
| Logical errors    | Program runs, but produces the wrong result   |



The simplest debugging technique is to print the value of a variable at different points in the program, to track the program state.

### Example: Linear search algorithm

The simplest way to find a value in a list: check each element one at a time from the beginning until the target is found.

What does this code do? 

````python
def linear_search(names, name_to_find):

    for name in names:
        if name == name_to_find:
            return True

    return False



robot_names = ["Martin", "Hemma", "Arthur", "Josh"]

print(linear_search(robot_names, "Arthur"))
````

Consider this version with a logical error introduced

- How many times does the loop execute?
- Does the function check every name?
- Why does it return False immediately?

````python
def linear_search(names, name_to_find):

    for name in names:
        if name != name_to_find:
            return False

    return True


robot_names = ["Martin", "Hemma", "Arthur", "Josh"]

print(linear_search(robot_names, "Arthur"))
````

### f strings 

Printing descriptively (i.e. with a string to explain what we are printing) can be really useful for debugging.

```python
print('Length of list', my_list, '=', len(my_list))
```

f strings avoid the need for mutiple arguments when using the `print` function. 

Including an f before the quotation marks of a string, allows variable names to be used between curly braces {} to insert the value of the variable into the string 


````python
my_list = [1,2,3]

print('Length of list', my_list, '=', len(my_list))

print(f'Length of {my_list} = {len(my_list)}')
````

### A systematic approach to debugging

(Steps 3 and 6 are new since last week)

1. **Identify the problem**<br>
   What is the program doing, and what did you expect it to do?
2. **Inspect the variables**<br>
   Which variables might be responsible for the unexpected behaviour?<br>
   Don't print everything! Think about which variables are relevant to the problem.<br>
3. **Add print statements** <br>
   Inspect their values before and after important operations.<br>
   Print intermediate results. The final output tells you that something went wrong; intermediate outputs help you find out where.<br>
   Labelling print statments with a string makes them much easier to interpret!
4. **Identify the underlying cause**
   Move print statements closer to the point where the unexpected behaviour occurs.
5. **Correct and test**<br>
   Fix the problem and check that the program works for several different inputs.<br>
   Test different inputs. A program that works for one example might still contain a bug.
6. **Clean up**   
   Remove unnecessary debugging print statements.

### Task 1: Calculating moving averages

The function should calculate the average of every group of three consecutive values.

Your tasks:
1. Use print() to inspect the windows being processed.
2. How many averages should the function return?
3. Why are the calculated averages incorrect?
4. Correct the function and test it with a different value of k.

````python
def moving_average(xs, k):
    averages = []

    for i in range(len(xs)-k):
        window = xs[i:i+k]
        average = sum(window) / len(xs)
        averages.append(average)

    return averages

data = [2, 4, 6, 8, 10, 12]

print(moving_average(data, 3))
````

### Task 2: Calculating the distance travelled by a robot

A robot records its position at regular intervals. 

The function should calculate the total distance travelled by adding the distances between consecutive positions.

However, the function produces an error.

Your tasks:
1. Run the function and read the error message.
2. Use print() to inspect the values being processed inside the loop.
3. Identify which line causes the error.
4. Explain why the error occurs.
5. Correct the function and test it with a different set of positions.

````python
# Answer = 12

def calculate_distance(positions):

    total_distance = 0

    for i in range(len(positions)):

        x1, y1 = positions[i]
        x2, y2 = positions[i+1]

        distance = ((x2 - x1)**2 + (y2 - y1)**2)**0.5

        total_distance = total_distance + distance

    return total_distance


robot_positions = [(0, 0), (3, 4), (6, 4), (6, 8)]

distance = calculate_distance(robot_positions)

print("Total distance:", distance)
````

### Task 3: Analysing student marks

Your tasks:
- What should the code do? 
- Identify the problem.
- Use print statements to investigate the values of question, student and total.
- Fix the function so it works for the given dataset.

````python
def calculate_averages(marks):

    averages = []

    for question in range(len(marks)):
        total = 0

        for student in marks:
            total += student[question]

        average = total / len(marks)
        averages.append(average)

    return averages


marks = [[6, 5, 5],
         [10, 4, 6],
         [7, 2, 5],
         [4, 3, 3]]

print(calculate_averages(marks))
````
