# `assert`

The `assert` statement is used to check that a condition is true during program execution. 

If the condition is True, the program continues normally. 

If it is False, Python raises an `AssertionError`.

````python
x = 10
2
assert x > 0
````

````python
x = -5
2
assert x > 0
````

````python
assert area_rectangle(2, 10) == 20 
````

````python
assert area_rectangle(-2, 10) >= 0
````

````python
area = area_rectangle(-2, 10)

assert area_rectangle(-2, 10) >= 0, f"Expected positive value"
````

# Choosing test cases

A good test suite includes...

__Representative__ (typical) inputs
```python
area_rectangle(5, 10)
area_rectangle(2, 3)
```

<br>

__Edge case__ inputs
```python
area_rectangle(0, 10)                # boundary value
area_rectangle(1, 1)                 # smallest positive dimensions
area_rectangle(-2, 10)               # invalid input
area_rectangle(1_000_000, 1_000_000) # very large values
```

# Example: Choosing test cases for a test suite

A student asks a third-party AI application:

>"Write a Python function that calculates the average speed from a list of speed measurements."

The AI generates:

```python
def average_speed(speeds):
    return sum(speeds) / len(speeds)
```

Let's write some tests for **representative** and **edge** case inputs

````python
def average_speed(speeds):
    return sum(speeds) / len(speeds)

def test_average_speed():
    assert average_speed([5, 6, 7]) == 6

def test_average_speed_ignores_direction():
    assert average_speed([5, -6, 7]) == 6

def test_ignore_invalid_speed():
    assert average_speed([5, None, 7]) == 6
````

# Example: Choosing test cases for a test suite

This example is taken from this week's lab exercises on lists

***
The function should return the neighbouring values around a chosen index in a list: the previous value, the current value, and the next value.

If the index is the last in the sequence, the next value should be the first in the sequence. 

If the index is the first in the sequence, the previous value should be the last in the sequence. 
***

In the exercise, you were asked to *Fix the function so that it can handle boundary index values (index=0 and index=4) and intermediate index values (e.g. index=2) correctly.* 

i.e. until it passed the tests using **representative** and **edge case** inputs. 

````python
def get_neighbours(values, index):
    previous = values[index - 1]
    current = values[index]

    # Check if index is tha last element in the list and express index as a negative value
    if index == len(values) - 1:
        index = -1

    next = values[index + 1]

    return [previous, current, next]
````

In the exercises, we tested the function by calling it using different arguments and printing the result
```python
get_neighbours(numbers, 2)
```

We can instead use a suite of test functions

````python
def test_get_neightbours_of_first_element():
    # Test edge case
    numbers = [10, 20, 30, 40, 50]
    neighbours = get_neighbours(numbers, 0) 
    assert neighbours == [50, 10, 20], f"Expected [50, 10, 20], got {neighbours}"

def test_get_neightbours_of_fifth_element():
    # Test edge case
    numbers = [10, 20, 30, 40, 50]
    neighbours = get_neighbours(numbers, 4) 
    assert neighbours == [40, 50, 10], f"Expected [40, 50, 10], got {neighbours}"

def test_get_neightbours_of_middle_element():
    # Test representative inputs
    numbers = [10, 20, 30, 40, 50]
    neighbours = get_neighbours(numbers, 2) 
    assert neighbours == [20, 30, 40], f"Expected [20, 30, 40], got {neighbours}"
    

test_get_neightbours_of_first_element()
test_get_neightbours_of_fifth_element()
test_get_neightbours_of_middle_element()
````

# Writing good tests



## 1. Test one thing at a time 
Avoid testing multiple unrelated behaviors in one test.

When deciding how to modularise your code using functions - a useful principle to apply is writing functions to allow one behaviour at a time to be tested.

## 2. Use descriptive test names 

````python
def test_rectangle_area_is_positive():
    assert area_rectangle(-5, 10) == 50
````

## 3. Structure tests consistently


__Arrange__: What input was used?

__Act__: What is being tested?  

__Assert__: What result was expected?



````python
def test_rectangle_area():
    # Arrange
    width = 5
    height = 10

    # Act
    area = area_rectangle(width, height)

    # Assert
    assert area == 50
````

## 4. Test both representative and edge cases

What inputs might a real user provide, and what should happen?


````python
assert area_rectangle(1, 1) == 1
assert area_rectangle(0, 10) == 0
````

## 5. Make assertions as specific as possible

````python
# More useful
assert area_rectangle(5, 10) == 50

# Less useful
assert area_rectangle(5, 10) > 0
````

## 6. Include descriptive failure messages

````python
area = area_rectangle(5, 10)
assert area == 50, f"Expected 50, got {area}"
````
