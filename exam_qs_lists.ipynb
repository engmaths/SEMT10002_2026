Q: This function is intended to return the sum of a list. Which change fixes the code?

```python
def total(values):
    s = 0
    for v in values:
        s += v

numbers = [2, 4, 6]

print(total(numbers))
```

A. Add `return s` at the end of the function

B. Add `print(s)` at the end of the function

C. Change `s += v` to `s = v`

D. Change `for v in values` to `for v in range(values)`

Answer: A

Q: What is printed?

```python
numbers = [1, 2, 3, 4]

result = []

for n in numbers:
    if n % 2 == 0:
        result.append(n * 10)

print(result)
```

A. [2, 4]

B. [10, 20, 30, 40]

C. [20, 40]

D. [1, 3]

Answer: C
