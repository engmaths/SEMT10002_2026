Wraparound boundary conditions:
- If the index is the last in the sequence, its upper neightbour is the first in the sequence 
- If the index is the first in the sequence, its lower neighbour is the last in the sequence

````python
def filter(xs, k):

    totals = []
    
    for i in range(len(xs)):

        total = sum(xs[i:i+k])
        
        totals.append(total)

    return totals

totals = filter([1,2,3,4,5,6,7,8,9], 3)
print(totals)

````

**Insertion sort** is a sorting algorithm that repeatedly takes takes the first item from the unsorted section of a list and compares this item backwards against the sorted section of the list, shifting larger elements to the right.

It can be more efficient than selection sort (shown in the notes for this week) because it only needs one comparison per element to confirm it is in place, whereas selection sort compares every element with all remaining unsorted elements.

````python
def insertion_sort(numbers):
    """
    Sorts a list of values in ascending order in-place
    """

    for i in range(1, len(numbers)):

        value = numbers[i] # Store value at current postion in list

        j = i-1            

        # Shift earlier values in the list to the right if they are greater than the stored value
        while j>=0 and numbers[j]>value:
            numbers[j+1] = numbers[j]
            j -= 1

        # When a lower value is reached, insert the stored value after it
        numbers[j+1] = value


my_list = [1,5,8,1,5,2]

insertion_sort(my_list)

print(my_list)
````

**Merge** can be used to combine two already-sorted lists into a single sorted list

````python
def merge(left, right):
    """
    Returns two already-sorted lists combined as a single sorted list
    """
    
    result = []
    i = 0
    j = 0
 
    while i < len(left) and j < len(right):
 
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1

        else:
            result.append(right[j])
            j += 1
 
    result += left[i:]
    result += right[j:]
 
    return result
````

**Merge sort** is an algorithm that first sorts small sections of data using *insertion sort* then merges them 

Example

`numbers = [38, 12, 27, 43, 9, 31, 18, 25]`

For this list, section length 4 gives lists

`[38, 12, 27, 43] [9, 31, 18, 25]`

Insertion sort operates on each section

`[12, 27, 38, 43] [9, 18, 25, 31]`

Finally they are merged

`[9, 12, 18, 25, 27, 31, 38, 43]`





````python
numbers = [38, 12, 27, 43, 9, 31, 18, 25, 3, 14]

def mergesort(numbers):
    """
    Sorts a list of values in ascending order in-place
    by sorting sections of data of maximum length 4 using 
    insertion sort, then merging increasingly large sections until a 
    single merged list achieved 
    """

    list_length = len(numbers)
    size = 4

    for start in range(0, list_length, size):

        section = numbers[start:start+size]

        # Insertion sort each section
        insertion_sort(section)

        # Replace section of original list with sorted section
        numbers[start:start+size] = section

    # Merge increasingly large sections until single merged list achieved 
    while size < list_length:
        for start in range(0, list_length, 2*size):
            left_section = numbers[start:start+size]
            right_section = numbers[start+size:start+2*size]

            merged = merge(left_section, right_section)

            numbers[start:start+len(merged)] = merged

        size *= 2



mergesort(numbers)

print(numbers)
````

Python's list method `sort` uses the **Timsort** algorithm. Timsort is similar to mergesort, but there's an important difference. 

Timsort doesn't simply divide the input into sections of fixed size. 

It identifies runs of ascending values in the data. For example:

`[1, 3, 5, 7, 9, 2, 4, 6, 8]`

already contains:

`[1, 3, 5, 7, 9]` and `[2, 4, 6, 8]`

Timsort idenitifies sorted sections ("runs"), then uses insertion sort only on unsorted sections, and finally merges all sections. 

(The real Timsort algorithm is more complicated, but this is the basic idea)
