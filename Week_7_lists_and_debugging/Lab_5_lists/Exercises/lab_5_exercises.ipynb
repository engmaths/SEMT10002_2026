A moving average filter is used in signal processing for smoothing data

It returns the average of every N consecutive values in a sequence

````python
def mystery(xs, k):
    
    best = sum(xs[:k])

    for i in range(1, len(xs) - k + 1):
        total = 0

        for j in range(i, i + k):
            total += xs[j]

        if total > best:
            best = total

    return best
````

````python
def get_neighbours(values, index):
    
    previous = values[index - 1]
    current = values[index]
    next = values[index + 1]

    return [previous, current, next]

numbers = [10, 20, 30, 40, 50]

# Testing function for intermediate index 2
result = get_neighbours(numbers, 2)
print(result)
````
