## A useful starting point

When you write Python code, things will go wrong. This is a normal part of programming.

Python provides information about what went wrong through **error messages**. These messages can initially look confusing, but they contain useful clues.

The aim of this seminar is to become better at interpreting those clues—not simply trying random changes until the error disappears.

## Error types

There are three broad categories of programming errors:

### 1. Syntax errors

These occur when Python cannot understand the structure of your code.

```python
x = 10
if x > 5
    print("Large number")
```

Python reports a `SyntaxError` because the `if` statement is missing a colon.
***

### 2. Runtime errors

These occur when Python understands your code but encounters a problem while executing it.

```python
numbers = [10, 20, 30]
print(numbers[5])
```

Python reports an `IndexError`: the code attempts to access an element that does not exist.
***

### 3. Logical errors

These occur when your code runs without producing an error message, but does not do what you intended.

```python
numbers = [2, 4, 6, 8]
total = 0

for number in numbers:
    total = number

print(total)
```

This prints `8`, rather than the sum of the numbers. The correction is:

```python
for number in numbers:
    total += number
```

**Important:** Not all programming errors produce error messages!

## Common Python error types

| Error | What it usually means |
|---|---|
| `SyntaxError` | Python cannot understand the code structure |
| `IndentationError` | Code is incorrectly indented |
| `NameError` | A variable or name has not been defined |
| `TypeError` | An operation is being applied to an inappropriate data type |
| `ValueError` | A function received an inappropriate value |
| `IndexError` | A list or sequence index is out of range |
| `KeyError` | A dictionary key does not exist |
| `AttributeError` | An object does not have the requested attribute or method |
| `ZeroDivisionError` | An attempt was made to divide by zero |
| `FileNotFoundError` | Python cannot find the requested file |

### Understanding the traceback

A traceback tells you how Python arrived at the point where the error occurred:

1) **Where did the error occur?** The filename, line number and function name help locate the relevant code.
2) **What was Python trying to execute?** The traceback displays the line of code being executed.
3) **What type of error occurred?** The final line gives the exception type and a short explanation.

The traceback will either look like the output of the cell above (.ipynb files) or like the text below (.py).<br> Both contain the same information. 

```text
Traceback (most recent call last):
  File "example.py", line 9, in <module>
    result = calculate_average(values)
  File "example.py", line 3, in calculate_average
    average = total / len(numbers)
ZeroDivisionError: division by zero
```

### A useful debugging technique

When reading a traceback:

1. Start at the bottom to identify the error type and message.
2. Look upwards to find the line(s) of code that triggered the error.
3. Read the surrounding code at that point in the program to understand why the problem occurred.
4. Check the values and types of the variables involved.

## A systematic approach to debugging 

When an error occurs, it is tempting to start changing code immediately. A better approach is to work systematically.


1. **Identify the problem** Read and understand the error message Identify the error type, line number and relevant code.
2. **Inspect the variables** Check the values and data types of the variables involved.
3. **Identify the underlying cause.** Ask what Python was trying to do and why it could not do it.
4. **Correct and test**. Make one change at a time. Avoid changing several things simultaneously.


Useful tools include:

```python
print(variable)
print(type(variable))
```
