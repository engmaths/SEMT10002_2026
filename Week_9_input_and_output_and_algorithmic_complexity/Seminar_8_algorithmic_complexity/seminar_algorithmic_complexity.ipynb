# Algorithmic Complexity

Algorithmic complexity is a measure of the growth of the computational resources required by an algorithm as a function of its input size $n$. 

This week we are moving beyond simply asking *"Does this algorithm work?"* to asking *"Does this algorithm work efficiently, and will it continue to do so as the problem gets bigger?"*...

We are primarily interested in two measures: 
- **Time complexity**: how the number of *operations* grows as the input gets bigger
- **Space complexity**: how the amount of *memory* required grows as the input gets bigger

# Big O

We use "Big O" notation to describe complexity. 

Big O describes the order of the growth of the time or space required by an algorithm as a function of the input size $n$

Different algorithms may solve the same problem but require very different amounts of time and space

- Compare how different algorithms will perform as the size of input data increases.
- Make informed decisions about which algorithm is most appropriate for a particular application.
- Understand the trade-offs between computational speed (time) and memory usage (space).

## How is algorithmic complexity calculated? : **Time complexity**


Algorithmic **time** complexity describes how the **number of operations** performed by an algorithm (the work done) changes as the input size $n$ increases.

Importantly, in this context, time is measured in number of operations *not* execution time in seconds. 


### How is the number of operations counted?

There are different types of operations used in a computer program. 

A **comparison** is an operation that checks whether two values satisfy a condition e.g.


e.g.
```Python
numbers[i] > numbers[j]

number == target

denominator != 0
```

We will introduce the concept of time complexity by counting *comparison operators* as a measure of time taken/work done by an algorithm:
- easy to identify
- often the dominant operation (e.g. sorting, searching) 

````python
numbers = [1,2,3,4]

if numbers[0] > numbers[1]: # COMPARISON
    print(True)
````

Now let's use counting comparison operators to identify:
- how the number of operations changes with the input size for some example algorithms 
- The complexity class the algorithm belongs to (e.g. $O(1)$, $O(n)$, $O(n^2)$,  $O(log\ n)$) 
- how an algorithm's performance is likely to compare to others as the input size increases 

## $O(n)$ - Linear time

Algorithms of complexity $O(n)$:
- Need to process every item in the input dataset
- Processing one item takes approximately the same amount of time, regardless of the dataset size.

The maximum number of comparisons is *proportional to* the **number of items** $n$

This is described as **linear complexity** because the number of operations grows *linearly* with the input size.

The growth is therefore of order $n$ and the complexity is written

$$O(n)$$

### Example algorithm: Linear Search, time complexity $O(n)$

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/linear_search.png?raw=true" width="30%">

Checks every item once until a target value is found
- Need to process every item in the input dataset
- Processing one item takes approximately the same amount of time, regardless of the dataset size.

````python
def linear_search(numbers, target):
    print('Input size', len(numbers))

    for i, number in enumerate(numbers):
        
        print(f'Comparison {i+1}: {number}=={target}')

        # COMPARISON
        if number == target: 
            return True

    return False
````

The number of comparisons is therefore equal to the number of items in the list

$$k=n$$


| Number of items $n$ | Max number of comparisons, $k$ |  
| :- | :- |  
| 1 | 1|  
| 5 | 5 |  
| 10 | 10 |  
| 100 | 100 |  
| 1,000 | 1,000 | 
| 10,000 | 10,000 | 





If there are **5 items** in the list to search, then the algorithm will perform a *maximum* of **5 comparisons**, $k$

````python
numbers = [4, 7, 2, 9, 5]
target = 5
outcome = linear_search(numbers, target)
print(f'{target} in list = {outcome}')
````

Depending on the input conditions (e.g. the target is not the final item in this case), the actual number of operations may be less than the maximum. 

````python
numbers = [4, 7, 2, 9, 5]
target = 2
outcome = linear_search(numbers, target)
print(f'{target} in list = {outcome}')
````

Suppose the list contains 1,000 elements.

| Situation | Number of comparisons |
|---|---:|
| Target is first | 1 |
| Target is in the middle | Approximately 500 |
| Target is last | 1,000 |
| Target is absent | 1,000 |

The *best case* is $O(1)$, but the *worst case* is $O(n)$.

Big O describes an *upper bound* on growth. 


### Example algorithm: Finding a maximum, time complexity $O(n)$

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/find_maximum.png?raw=true" width="70%">

Checks every item to find the largest in the series
- Need to process every item in the input dataset
- Processing one item takes approximately the same amount of time, regardless of the dataset size.

````python
def find_largest(numbers):

    print('Input size', len(numbers))
    
    largest = numbers[0]

    for i, number in enumerate(numbers):
        
        print(f'Comparison {i+1}: {number} > {largest}')

        # COMPARISON
        if number > largest:
            largest = number

    return largest
````

````python
numbers = [4, 7, 2, 9, 5]
print(find_largest(numbers), 'is largest number')
````

Again, the number of comparisons is equal to the number of items in the list

$$k=n$$


Notice, that becuase `largest` is initialised as `numbers[0]`, we can actually reduce the number of operations to

$$k = n-1$$

````python
def find_largest(numbers):

    print('Input size', len(numbers))
    
    largest = numbers[0]

    for i, number in enumerate(numbers[:-1]):
        
        print(f'Comparison {i+1}: {number} > {largest}')

        # COMPARISON
        if number > largest:
            largest = number

    return largest
````

````python
numbers = [4, 7, 2, 9, 5]
print(find_largest(numbers), 'is largest number')
````

#### Why is this still linear complexity?

As the input size increases the number of comparisons increases approximately proportional to $n$.

Small order terms become negligable as the input size becomes large (the accuarcy of the approximation increases).

| Input size ($n$) | Total comparisons ($k=n-1$) |
|---:|---:|
| 1 | 0 |
| 2 | 1 |
| 5 | 4 |
| 10 | 9 |
| 100 | 99 |
| 1,000 | 999 |

To reflect this, when calculating the complexity, **smaller-order** terms are ignored. 

**Constant multipliers** are also ignored becuase Big O notation indicates the *order* of the growth of the work done by the algorithm as a funtion of the input size $n$, not the *actual* work done.  

So an algorithm with number of comparisons $k = n-1$ has linear complexity $O(n)$

## $O(n^2)$ - Quadratic time 

Algorithms of complexity $O(n^2)$:
- Need to perform up to $n$ operations (e.g. comparisons) per item
- Consequently, the total number of operations grows approximately with the square of the input size, $n^2$

This is described as **quadratic complexity** because the number of operations grows *quadratically* with the input size.

The growth is therefore of order $n^2$ and the complexity is written

$$O(n^2)$$

### Example algorithm: Finding repeat values, time complexity $O(n^2)$

Each item is compared with every other item.

If two items are the same, and their index is different (i.e. not black squares), a repeat value is recorded (yellow squares).

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/find_repeats_v1.png?raw=true" width="30%">

This is achieved using two loops

For each iteration of the outer loop $n$, there are $n$ iterations of the inner loop

Therefore there are $n^2$ comparisons made

(Notice that the list of repat values is converted to a set so that it contains unique values only)

````python
robot_names = ["Josh", "Hemma", "Martin", "Josh"]

def find_repeats_v1(items):

    repeat_values = []

    # N outer loop iterations
    for n, name_a in enumerate(items):

        # N inner loop iterations
        for m, name_b in enumerate(items):

            # COMPARISON
            if (name_a == name_b and n!=m): 
                repeat_values.append(name_a)
                
    return set(repeat_values)

find_repeats_v1(robot_names)
````

Now consider a slightly different way to approach this problem. 

We compare each item in the list to the items to its right *only*. 

Each items has already been compared to the items to its left, so we don't need to repeat the operation

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/find_repeats_v2_.png?raw=true" width="40%">



````python
robot_names = ["Josh", "Hemma", "Martin", "Josh"]

def find_repeats_v2(items):

    repeat_values = []

    # N outer loop iterations
    for i, name_a in enumerate(items[:-1]):

        # N-i inner loop iterations
        for name_b in items[i+1:]:
            if (name_a == name_b): 
                repeat_values.append(name_a)

    return set(repeat_values)

find_repeats_v2(robot_names)
````

The total number of comparisons forms a triangular number series:

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/triangular_numbers.png?raw=true" width="35%">

$k = (n-1) + (n-2) + ...  + 2 + 1$

If we add the series to the same series written in reverse, we get:

$2k = n + n + n + ... + n$

(There are $(n-1)$ terms each equal to $n$)

$2k = n (n-1)$

So we can simplify the number of comparisons to:

$$k = \frac{n(n-1)}{2} = \frac{n^2-n}{2}$$

So what is the complexity? 

Remeber that in Big O notation, **constant multipliers** and **smaller-order** terms are ignored 

So an algorithm with number of comparisons $k = \frac{n^2-n}{2}$ has linear complexity $O(n^2)$, as $n^2$ is the dominant term



So if we compare the two quadratic complexity algorithms implemented, their time complexity (order of growth with $n$) is the same category, $O(n^2)$ but `find_repeats_v2` requires approximately half as much work as `find_repeats_v1` for a given input size $n$

| $n$ | `find_repeats_v2`<br>$k=\frac{n^2-n}{2}$ |  `find_repeats_v1`<br>$k=n^2$ |
|:---|:---|:---|
| 2 | 1 | 4 |
| 5 | 10 | 25 |
| 10 | 45 | 100 |
| 20 | 190 | 400 |
| 50 | 1,225 | 2,500 |
| 100 | 4,950 | 10,000 |
| 500 | 124,750 | 250,000 |
| 1,000 | 499,500 | 1,000,000 |
| 10,000 | 49,995,000 | 100,000,000 |

| Algorithm / task | Example | Why is it $O(n^2)$? |
|---|---|---|
| Comparing items to all other items | Finding pairs of with matching characteristics | Each item is compared with every other item. |
| Duplicate detection (nested loops) | Checking whether a list contains duplicate values | Each item may be compared against every other item. |
| Selection sort | Finding the smallest remaining item and moving it into position | Repeatedly searches through the remaining unsorted items. |
| Insertion sort | Inserting each new item into its correct position in a sorted section | Repeatedly searches through the sorted items.  |
| Bubble sort | Sorting a list by repeatedly comparing adjacent elements | Repeatedly traverses the list, making comparisons between elements. |

## $O(n^x)$ - Polynomial time 

Nested loops can tell us something about the complexity of an algorithm

The time complexity increases exponentially with the number of nested loops iterating over $n$ items

- One loop over $n$ items → $O(n)$.
- Two nested loops, each iterating over $n$ items → $O(n^2)$.
- Three nested loops, each iterating over $n$ items → $O(n^3)$.







### Example algorithm: Generate all $x$-digit combinations, time complexity $O(n^x)$

Find all combinations of $x=$ 3 digits, where each digit has $n=$ 2 possible values 

How does it effect the number of operations:
- if $x$ increases?
- if $n$ increases?

$x=3$ so the time complexity is $O(n^3)$ 

````python
combinations = []

input = [0,1]

for a in input:
    for b in input:
        for c in input:
            combo = [a, b, c]
            combinations.append(combo)

print('Number of operations =', len(combinations))

for i in combinations:
    print(i)
````

## $O(\log n)$ - Logarithmic time 

Algorithms of complexity $O(\log n)$:
- **Do not need to check every item:** <br>Instead at each step they check one item, then discard some (typically half) remaining possibilities
- **Logarithmic relationship between input size $n$ and number of operations $k$:** <br>e.g. Multiplying $n$ by a factor of 2, increases $k$ by  1

If we halve the remaining values at each step, the number of operations $k$ needed to reduce the remaining values from $n$ items to 1 item is:

$k = \log_2(n)$

| Items, $n$ | Operations, $k$ |
|:--|:--|
| 1 | 0 |
| 2 | 1 |
| 4 | 2 |
| 8 | 3 |
| 16 | 4 |

This is described as **logarithmic complexity** because the number of operations grows *logarithmically* with the input size.

The growth of the algorithm is therefore of order $\log (n)$ and the complexity is written

$$O(\log n)$$



Let's illustrate this with an example...

### Example algorithm: Binary search, time complexity $O(\log n)$

This algorithm works on a *sorted* list.

It repeatedly:
- checks if the middle item is the target value (comparison)
- discards half the remaining items if it is not (lower half if target > middle, upper half if target < middle)

It stops when either:
- the target is found
- the whole list has been checked (the upper and lower bound overlap)

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/binary_search.png?raw=true" width="35%">



````python
def binary_search(numbers, target):

    loop_counter = 1

    # Upper and lower bound of range to check
    low = 0
    high = len(numbers)-1

    # When the upper and lower bound overlap, we have checked the whole list
    while low <= high:

        # Find the middle index
        middle = (low + high) // 2
        print('Middle =', numbers[middle])
    
        # ----------------------------
        # COMPARISONS
        print(f'Comparison {loop_counter}: {numbers[middle]} == {target}')
        loop_counter += 1

        if numbers[middle] == target:   
            return True
        elif numbers[middle] < target:
            low = middle + 1    # Increase lower bound 
        else:
            high = middle - 1   # Reduce upper bound

        print()
        # -----------------------------

    return False

````

The maximum number of comparisons is:

$$k = \log_2(n) + 1$$

($\log_2(n)$ comparisons before each operation to reduce the data set from $n$ items to 1 item<br>
$1$ additional comparison is required to check if the final remaining item is equal to the target.)

| Items, $n$ |  Comparisons<br> $k = \log_2(n) + 1$ |
|---:|---:|
| 1 | 1 |
| 2 | 2 |
| 4 | 3 |
| 8 |  4 |
| 10 |  4 |
| 16 |  5 |
| 20 |  5 |
| 32 |  6 |
| 64 |  7 |
| 1,024 |  11 |
| 1,048,576 |  21 |

Non-integer values of $\log_2(10)$ are rounded down to give the number of operations to reduce the data set to 1 item, as this must be a whole number.

For example if $n=10$

$k = \lfloor \log_2(10) \rfloor + 1 = \lfloor 3.322 \rfloor + 1 = 4$ 



Consider a sorted list of 10 numbers

Check if:
- 2 is in the list
- 5 is in the list
- 7 is in the list
- 11 is in the list

````python
numbers = [-5, -2, 0, 1, 2, 4, 5, 6, 7, 10]

target = 7
outcome = binary_search(numbers, target)

print()
print(f'{target} in list = {outcome}')
````

Remeber that in Big O notation, **constant multipliers** and **smaller-order** terms are ignored

So an algorithm with number of comparisons $k = \lfloor \log_2(10) \rfloor + 1$ has linear complexity $$O(\log (n))$$

So we can see, that for an already sorted list, binary search is far more efficient than linear search. 

| Number of items<br> $n$ | Linear search<br> $O(n)$ | Binary search<br>  $O(\log_2 n)$ |
|:---|:---|:---|
| 10 | 10 | 4 |
| 100 | 100 | 7 |
| 1,000 | 1,000 | 10 |
| 10,000 | 10,000 | 14 |
| 100,000 | 100,000 | 17 |
| 1,000,000 | 1,000,000 | 20 |
| 10,000,000 | 10,000,000 | 24 |
| 100,000,000 | 100,000,000 | 27 |
| 1,000,000,000 | 1,000,000,000 | 30 |

## $O(1)$ - Constant time 

Algorithms of constant complexity $O(1)$:
- **A fixed number of operations:** <br>They perform approximately the same number of operations regardless of the size of the input dataset.
- **No need to traverse the dataset:** <br>They do not need to examine every item or search through the input.
- **Direct access to information:** <br>They often use an index, key or previously stored information to access the required data.
- **Input size does not affect the number of operations:** <br>e.g. Doubling the dataset size does not increase the number of operations.

## How is algorithmic complexity calculated? : **Space complexity**


Algorithmic **space** complexity describes how the **amount of memory** required by an algorithm changes as the input size $n$ increases.

Space complexity usually refers to *additional memory* used by the algorithm, rather than the memory already occupied by the input.
 


Just as with time complexity, Big O is used to describe space complexity:

| Space complexity | Example | Why? |
|---|---|---|
| $O(1)$ | Finding the largest number in a list | Only a fixed number of additional variables are needed, regardless of the list size. |
| $O(n)$ | Creating a copy of a list | The new list requires space proportional to the number of input items. |
| $O(n^2)$ | Creating an $n \times n$ matrix | The matrix contains $n^2$ elements. |
| $O(\log n)$ | Storing the remaining unseacrhed items at each step of a binary search | Each iteration halves the remaining items. |

Space complexity becomes a bit difficult to show through elementary examples as it requires a knowldge of underlying mechanisms of computer memory that is outside the scope of this unit. 

But we can get a general understanding through some simple code examples:

## $O(1)$ - Constant space

### Example algorithm: Finding a maximum $O(1)$

A function we saw earlier (time complexity $O(n)$) that finds the largest number in a list.

Only one additional variable (`largest`) is needed, regardless of how many numbers are in the list.

````python
def find_largest(numbers):
    
    largest = numbers[0]

    for number in numbers[:-1]:

        if number > largest:
            largest = number

    return largest

numbers = [4, 7, 2, 9, 5]
find_largest(numbers)
````

### Example algorithm: Linear search, space complexity $O(1)$

A function we saw earlier (time complexity $O(n)$) that looks for a target item in a list.

No additional variables are needed, regardless of how many numbers are in the list - this is also a constant value.

````python
def linear_search(numbers, target):
    print('Input size', len(numbers))

    for i, number in enumerate(numbers):
        
        print(f'Comparison {i+1}: {number}=={target}')

        # COMPARISON
        if number == target: 
            return True

    return False
````

### Example algorithm: Binary search, space complexity $O(1)$

A function we saw earlier (time complexity $O(log n)$) that searches for a value in a sorted list

Only three additional variables (`low`, `middle`, `high`) are needed, regardless of how many numbers are in the list.

````python
def binary_search(numbers, target):

    loop_counter = 1

    # Upper and lower bound of range to check
    low = 0
    high = len(numbers)-1

    # When the upper and lower bound overlap, we have checked the whole list
    while low <= high:

        # Find the middle index
        middle = (low + high) // 2
        print('Middle =', numbers[middle])
    
        # ----------------------------
        # COMPARISONS
        print(f'Comparison {loop_counter}: {numbers[middle]} == {target}')
        loop_counter += 1

        if numbers[middle] == target:   
            return True
        elif numbers[middle] < target:
            low = middle + 1    # Increase lower bound 
        else:
            high = middle - 1   # Reduce upper bound

        print()
        # -----------------------------

    return False

````

## $O(n)$ - Linear space

### Example algorithm: Creating a copy of a list. $O(n)$

Every additional input item requires another element in the new list.

````python
def copy_list(numbers):
    new_list = []

    for number in numbers:
        new_list.append(number)

    return new_list

numbers = [4, 7, 2, 9, 5]
copy_list(numbers)
````


### Example algorithm: Generating a dictionary, space complexity $O(n)$

Creating a dictionary containing the square of each number.

````python
def create_squares(numbers):
    squares = {}

    for number in numbers:
        squares[number] = number ** 2

    return squares

numbers = [4, 7, 2, 9, 5]
create_squares(numbers)
````

## $O(n^2)$ - Quadratic space 

### Example algorithm: Creating a multiplication table, space complexity $O(n^2)$

For an input of size $n$, the function creates an $n \times n$ matrix.

(Note that the nested loops in the multiplication table are not what make its space complexity $O(n^2)$. It is the fact that we store $n^2$ elements in memory.)



````python
def multiplication_table(n):
    table = []

    for i in range(1, n+1):
        row = []

        for j in range(1, n+1):
            row.append(i * j)

        table.append(row)

    return table

multiplication_table(5)
````

### Example algorithm: Find repeat values, space complexity $O(n^2)$

A function we saw earlier (time complexity $O(n^2)$) that looks for repeat values in a list.

The variable `repeat values` has a maximum size of $n$ if all items are the same, as each comparison will result in a repeat value being stored   

````python
robot_names = ["Josh", "Hemma", "Martin", "Josh"]

def find_repeats_v2(items):

    repeat_values = []

    # N outer loop iterations
    for i, name_a in enumerate(items[:-1]):

        # N-i inner loop iterations
        for name_b in items[i+1:]:
            if (name_a == name_b): 
                repeat_values.append(name_a)

    return set(repeat_values)

find_repeats_v2(robot_names)
````

There is often a *trade-off* between space efficiency and time efficieny of an algorithm and so the most suitable algorithm often depends on the application. 



| Algorithm / approach | Time complexity | Space complexity | Explanation / trade-off |
|---|---|---|---|
| Accessing a list element by index | $O(1)$ | $O(1)$ | Directly accesses an element without searching. |
| Accessing a dictionary value by key | $O(1)$ average | $O(n)$ | Uses a hash table for fast lookup, at the cost of additional memory. |
| Binary search | $O(\log n)$ | $O(1)$ | Repeatedly halves the search space. Requires an already sorted list. |
| Linear search | $O(n)$ | $O(1)$ | May need to check every item, but requires very little additional memory. |
| Finding the largest value | $O(n)$ | $O(1)$ | Examines every item while keeping track of the largest value found. |
| Copying a list | $O(n)$ | $O(n)$ | Creates a new list containing every input element. |
| Creating a dictionary of squared values | $O(n)$ | $O(n)$ | Creates a dictionary entry for each input number. |
| Merge sort | $O(n\log n)$ | $O(n)$ | Uses additional memory for intermediate lists to achieve faster sorting. |
| Finding repeated values (nested loops) | $O(n^2)$ | $O(n^2)$ | Compares every item with every other item. |
| Finding repeated values (optimised nested loops) | $O(n^2)$ | $O(n^2)$ | Avoids comparing items twice, reducing the number of comparisons but retaining quadratic growth. |
| Selection sort | $O(n^2)$ | $O(1)$ | Repeatedly finds the smallest remaining item and moves it into position. |
| Insertion sort | $O(n^2)$ | $O(1)$ | Sorts within the original list, requiring very little additional memory. |
| Bubble sort | $O(n^2)$ | $O(1)$ | Repeatedly compares and swaps adjacent elements within the original list. |
| Creating a multiplication table | $O(n^2)$ | $O(n^2)$ | Creates and stores an $n \times n$ matrix. |
| Generating all combinations of $x$ digits | $O(n^x)$ | $O(n^x)$ | Generates and stores every possible combination. |

## Cheat sheet

This table decribes the identifying features associted with the compexity classes we have studied today

| Big O classification | Time complexity: identifying features | Space complexity: identifying features |
|---|---|---|
| $O(1)$ Constant | - Does not need to process every item in the input dataset.<br>- Performs a fixed number of operations, regardless of input size. | - Uses a fixed amount of additional memory.<br>- The number of additional variables or storage locations does not increase with the input dataset size. |
| $O(n)$ Linear | - Needs to process every item in the input dataset.<br>- Processing one item takes approximately the same amount of time, regardless of the dataset size.<br>| - Additional memory grows in proportion to the input dataset size.<br>- Typically creates a new list, dictionary or other data structure containing one additional element for each input item.<br>|
| $O(n^x)$ Polynomial | - The number of operations grows as a power of the input size.<br>- Often involves nested loops that iterate over the input dataset.\*\* | - Additional memory grows as a power of the input size.<br>- Often involves creating multidimensional data structures. |
| $O(\log n)$ Logarithmic | - Each operation significantly reduces the remaining problem size.<br>- The number of operations increases slowly as the input dataset grows. | - The amount of additional memory grows logarithmically with the input size. |

 \*\* Nested loops do not automatically mean quadratic complexity. We need to consider how many times the inner loop actually executes and how this changes as the input grows.

### Exercise 2

Selection sort:
- iterates through a list
- repeatedly finds the smallest element in the unsorted part of a list
- swaps it with element at the index of the loop counter
- continues until the entire list is sorted.

Analyse the complexity of this algorithm:

- How many times is the comparison applied for a given input size?
- How does this scale with the input size? 
- Which of the examples we have studeied today does this look most similar to? 
- What do you think the Big O complexity of the algorithm is? 


````python
def selection_sort(numbers):
    """
    Sorts a list of values in ascending order in-place
    """

    # Iterate through the list
    for i in range(len(numbers)):

        # Initialise index of the miniumum value as the loop counter
        min_index = i

        # Update index of the miniumum value as smallest value in unsorted part
        for j in range(i + 1, len(numbers)):
            if numbers[j] < numbers[min_index]:
                min_index = j
        
        # Swap smallest value and value at index of loop counter
        numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

numbers = [1,2,5,7,2,5,6]
selection_sort(numbers)
print(numbers)
````
