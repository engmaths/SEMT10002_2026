# Algorithmic Complexity

Algorithmic complexity is a measure of the computational resources required by an algorithm change as a function of its input size $n$. 

We are primarily interested in two measures: 
- **Time complexity**: how the number of operations grows as the input gets bigger
- **Space complexity**: how the amount of memory required grows as the input gets bigger

## Why should we care about algorithmic complexity?

Different algorithms can solve the same problem but require very different amounts of time and memory. 

An algorithm that works fine for a small dataset might become impractically slow or memory-intensive when the dataset grows.

By studying algorithmic complexity, we can:
- Compare different algorithms and understand their relative efficiency.
- Predict how algorithms will perform as the size of a problem increases.
- Make informed decisions about which algorithm is most appropriate for a particular application.
- Understand the trade-offs between computational speed and memory usage.

Ultimately, algorithmic complexity helps us move beyond simply asking *"Does this algorithm work?"* to asking *"Does this algorithm work efficiently, and will it continue to do so as the problem gets bigger?"*

## How is algorithmic complexity calculated?

### Time complexity 

Algorithmic time complexity is calculated by examining how the number of **operations** performed by an algorithm changes as the size of its input increases.

There are differnt types of operations used in a computer program. 

A **comparison** is an operation that checks whether two values satisfy a condition e.g.


Counting the number of **comparisons** an algorithm makes is a useful way to introduce the concept of complexity because:

### Example: Linear Search $O(n)$

Suppose we want to find a particular value in a list.

````python
def linear_search(numbers, target):
    for number in numbers:
        if number == target: # Comparison
            return True

    return False
````

If there are **5 items** in the list to search, then the algorithm will perform a *maximum* of **5 comparisons**

````python
numbers = [4, 7, 2, 9, 5]
target = 5
````

| :-: | :-: | :-: | 
| 1 | 4==5 | False | 
| 2 | 7==5 | False |

| 5 | 5==5 | True |

The maximum number of comparisons is equal to the number of items $n$

This is described as **linear complexity** because the number of operations grows *linearly* with the input size.

The growth is therefore of order $n$ and the complexity is written

$$O(n)$$

### Example: Binary search $O(\log n)$

Now consider binary search. This algorithm works on a sorted list.

Rather than checking every item, it repeatedly divides the search area in half.

````python
def binary_search(numbers, target):
    low = 0
    high = len(numbers) - 1

    while low <= high:
        middle = (low + high) // 2

        # ----------------------------
        # Comparison
        if numbers[middle] == target:   
            return True
        elif numbers[middle] < target:  
            low = middle + 1
        else:
            high = middle - 1
        # -----------------------------

    return False

binary_search(numbers, 38)
````

| 38 | 38==38 | True |  |

The key idea is that each comparison approximately halves the remaining search area.

We can calculate the number of comparisons by repeatedly dividing the list size by two.

For a list with 16 items 

| Number of comparisons  | Items remaining |

If we start with $n$ items, the number of items remaining, $j$, after $k$ comparisons can be expressed as

$$\frac{n}{2^k} = j$$

Letting $j=1$

$\frac{n}{2^k} = 1$

Rearrange 

$n = 2^k$

*To what power $k$ must we raise $2$ to get $n$?*

The maximum number of comparisons $k$ is equal to $ \log_2(n)$

This is described as **logarithmic complexity** because the number of operations grows *logarithmically* with the input size.

The growth of the algorithm is therefore of order $\log (n)$ and the complexity is written

$$O(\log n)$$

Binary search is logarithmic because every comparison divides the remaining search area by a factor (of 2 in this case).

The outer loop runs three times.

For every iteration of the outer loop, the inner loop also runs three times.

The maximum number of comparisons is equal to $n^2$

This is described as **quadratic complexity** because the number of operations grows *quadratically* with the input size.

The growth is therefore of order $n^2$ and the complexity is written

$$O(n^2)$$

````python
def selection_sort(numbers):
    """
    Sorts a list of values in ascending order in-place
    """

    for i in range(len(numbers)):

        '''
        Initialise the index of the miniumum unsorted value as 
        the current position in the list
        '''
        min_index = i

        ''' 
        Update this index with the index of the smallest value 
        to the right of the current position
        '''
        for j in range(i + 1, len(numbers)):
            if numbers[j] < numbers[min_index]:  # COMPARISON
                min_index = j
        
        '''
        Swap the smallest value to the right of the current position 
        with the value at the current position
        '''
        numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

numbers = [1,2,5,7,2,5,6]
selection_sort(numbers)
print(numbers)
````

**How does the amount of work an algorithm needs to do change as the size of the problem increases?**

This is the central idea behind algorithmic complexity.

Algorithmic complexity describes how the computational resources required by an algorithm change as the input size increases.

We are particularly interested in two resources:

**Time complexity:** How does the *number of operations* performed by an algorithm grow as the input gets larger?

**Space complexity:** How does the *amount of additional memory* required by an algorithm grow as the input gets larger?

We are not initially interested in measuring actual time in seconds. This depends on the computer, processor, programming language and implementation. Instead, we want to understand how the algorithm itself scales.

### Example: Two sorting algorithms

Suppose two students write different algorithms to sort a list.

- Algorithm A takes 1 second to sort 100 numbers.
- Algorithm B takes 2 seconds to sort 100 numbers.

At first glance, Algorithm A looks better.

However...

- Algorithm A takes 100 seconds to sort 10,000 numbers.
- Algorithm B takes 5 seconds to sort 10,000 numbers.

Which algorithm would you rather use?

The fastest algorithm for a small input is not necessarily the fastest algorithm for a large input.

**The aim of complexity analysis is to understand how algorithms scale, rather than simply how quickly they run on one particular example.**

## Measuring the work performed by an algorithm

### Counting operations

````python
def print_numbers(numbers):

    for number in numbers:
        print(number)
````

3. How would you describe the relationship between input size and the number of operations?

We can describe this relationship as:

$\text{Number of operations} \propto n$

where $n$ represents the number of elements in the input.

This is called **linear complexity**.

### Example algorithm: Searching for the maximum 

````python
def find_maximum(numbers):

    maximum = numbers[0]

    for number in numbers:
        if number > maximum:
            maximum = number

    return maximum
````

The function finds the largest number in a list.

The algorithm:
1. Stores the first number as the current maximum.
2. Iterates through the list.
3. Compares each number with the current maximum.
4. Updates the maximum when necessary.

The algorithm needs to inspect every element in the list.

If the list contains 5 elements, it performs approximately 5 comparisons. 

If it contains 100 elements, it performs approximately 100 comparisons.

We can describe this relationship as:

$$\text{Number of operations} \propto n$$

where $n$ represents the number of elements in the input.

**Key idea:** When an algorithm processes every element once, its *time complexity* is generally linear.

## Big O notation: Describing how algorithms scale

We use **Big O notation** to describe how computational work grows as the input size increases.

For this seminar, we will concentrate on four important complexity classes.

| Complexity | Name | Example algorithm|
|---|---|---|
| $O(1)$ | Constant | Accessing an element at a known list index |
| $O(\log n)$ | Logarithmic | Binary search |
| $O(n)$ | Linear | Linear search |
| $O(n^2)$ | Quadratic | Selection sort; insertion sort in its worst case |

### Understanding the difference

The following table gives illustrative operation counts. 

These are not exact predictions for every algorithm.

| Input size $n$ | $O(1)$ | $O(\log_2 n)$ | $O(n)$ | $O(n^2)$ |
|---:|---:|---:|---:|---:|
| 10 | 1 | 4 | 10 | 100 |
| 100 | 1 | 7 | 100 | 10,000 |
| 1,000 | 1 | 10 | 1,000 | 1,000,000 |
| 10,000 | 1 | 14 | 10,000 | 100,000,000 |

Notice what happens when the input size increases from 1,000 to 10,000:

- $O(1)$: no increase.
- $O(\log n)$: increases by only a few operations.
- $O(n)$: increases tenfold.
- $O(n^2)$: increases one hundredfold.

This explains why the choice of algorithm becomes increasingly important as the problem gets larger.

## 4. Comparing searching algorithms

Searching provides a useful way of understanding algorithmic complexity because we have encountered two different approaches.

### 4.1 Linear search: \(O(n)\)

Linear search checks each element in turn until it finds the target or reaches the end of the list.

```python
def linear_search(numbers, target):

    for i in range(len(numbers)):

        if numbers[i] == target:
            return i

    return -1
```

For example:

```python
numbers = [12, 8, 35, 19, 42, 7, 51]
```

Searching for 42 requires five comparisons.

If the target is not present, the algorithm must inspect every element before concluding that the value is missing.

- **Best case:** \(O(1)\), if the target is the first element.
- **Worst case:** \(O(n)\), if the target is last or absent.
- **Average case:** \(O(n)\), assuming a typical unsorted list.

### 4.2 Binary search: \(O(\log n)\)

Binary search takes a different approach. Rather than checking every element, it repeatedly divides the search space in half.

**Important:** Binary search only works correctly when the data is sorted.

Consider:

```python
numbers = [3, 7, 12, 19, 25, 35, 42, 51, 60]
```

Suppose we are searching for 42:

1. Look at the middle element: 25.
2. 42 is greater than 25, so discard the lower half.
3. Look at the middle of the remaining section: 42.
4. The target is found.

Each comparison eliminates approximately half of the remaining possibilities.

For an input of size \(n\), the number of halvings is approximately:

\[
\log_2(n)
\]

| Number of elements | Approximate halvings |
|---:|---:|
| 8 | 3 |
| 16 | 4 |
| 32 | 5 |
| 64 | 6 |
| 128 | 7 |
| 1,024 | 10 |
| 1,000,000 | 20 |

This is why binary search has logarithmic complexity.

````python
def binary_search(numbers, target):
    low = 0
    high = len(numbers) - 1

    while low <= high:
        middle = (low + high) // 2

        if numbers[middle] == target:
            return middle
        elif numbers[middle] < target:
            low = middle + 1
        else:
            high = middle - 1

    return -1


numbers = [3, 7, 12, 19, 25, 35, 42, 51, 60]
print(binary_search(numbers, 42))
print(binary_search(numbers, 20))
````

## 5. Comparing sorting algorithms

Different sorting algorithms can produce exactly the same result but require very different amounts of work.

We will compare:

- Selection sort
- Insertion sort
- Bubble sort
- Merge sort

### 5.1 Selection sort: \(O(n^2)\)

Selection sort repeatedly finds the smallest element in the unsorted part of the list and moves it into the correct position.

For example:

```python
numbers = [5, 3, 8, 1, 4]
```

The algorithm:

1. Finds the smallest element (1) and moves it to the first position.
2. Finds the smallest remaining element (3) and moves it to the second position.
3. Repeats until the list is sorted.

The number of comparisons is:

\[
(n-1)+(n-2)+\ldots+2+1
\]

This sum is:

\[
\frac{n(n-1)}{2} = \frac{n^2-n}{2}
\]

For large \(n\), the \(n^2\) term dominates, so we describe the complexity as \(O(n^2)\).

If we double the input size, the number of comparisons increases approximately fourfold.

### 5.2 Insertion sort: \(O(n^2)\)

Insertion sort builds a sorted section of the list, inserting each new element into its correct position.

Think about sorting a hand of playing cards:

1. Pick up the next card.
2. Compare it with the cards already sorted.
3. Move larger cards to the right.
4. Insert the new card into the gap.

In the worst case, the list is in reverse order. Every new element must be moved past all the elements that came before it.

The number of comparisons or movements is approximately:

\[
1+2+3+\ldots+(n-1)
\]

Again, this gives \(O(n^2)\).

However, if the list is already sorted, each new element only needs to be checked against the previous element. Its best-case complexity is therefore \(O(n)\).

This makes insertion sort particularly useful for small or nearly sorted datasets.

### 5.3 Bubble sort: \(O(n^2)\)

Bubble sort repeatedly compares adjacent elements and swaps them if they are in the wrong order.

After one pass through the list, the largest element has moved to the end. The algorithm repeats until the list is sorted.

In the worst case, bubble sort performs approximately \(n^2\) operations.

An optimised implementation can stop early if no swaps are made during a pass. If the list is already sorted, this gives a best-case complexity of \(O(n)\).

### 5.4 Merge sort: \(O(n \log n)\)

Merge sort uses a strategy called **divide and conquer**.

The algorithm:

1. Divides the list into two halves.
2. Recursively divides each half until individual elements remain.
3. Merges the smaller lists back together in sorted order.

There are two important observations:

- The list is divided approximately \(\log_2(n)\) times.
- At each level, merging the lists requires processing approximately \(n\) elements.

Therefore:

\[
O(n)\times O(\log n) = O(n\log n)
\]

This grows much more slowly than \(O(n^2)\).

### 5.5 Comparing the sorting algorithms

| Algorithm | Best case | Average case | Worst case |
|---|---:|---:|---:|
| Selection sort | \(O(n^2)\) | \(O(n^2)\) | \(O(n^2)\) |
| Insertion sort | \(O(n)\) | \(O(n^2)\) | \(O(n^2)\) |
| Bubble sort* | \(O(n)\) | \(O(n^2)\) | \(O(n^2)\) |
| Merge sort | \(O(n\log n)\) | \(O(n\log n)\) | \(O(n\log n)\) |

*Assuming bubble sort uses an early-termination check.*

The algorithms respond differently to the input:

- Selection sort performs approximately the same number of comparisons regardless of whether the input is sorted.
- Insertion sort performs much less work when the input is already sorted.
- Merge sort maintains approximately the same complexity regardless of the initial ordering.

## 7. Best case, worst case and average case

The same algorithm can perform differently depending on the input.

- **Best case:** The input arrangement that allows the algorithm to perform the least work.
- **Worst case:** The input arrangement that requires the algorithm to perform the most work.
- **Average case:** The expected amount of work across a range of possible inputs, under assumptions about their distribution.

Consider linear search:

```python
def linear_search(numbers, target):

    for number in numbers:

        if number == target:
            return True

    return False
```

Suppose the list contains 1,000 elements.

| Situation | Number of comparisons |
|---|---:|
| Target is first | 1 |
| Target is in the middle | Approximately 500 |
| Target is last | 1,000 |
| Target is absent | 1,000 |

The best case is \(O(1)\), but the worst case is \(O(n)\).

**Important:** Big O does not automatically mean worst case. Big O describes an upper bound on growth. We can use it to describe best-case, average-case or worst-case behaviour.

For introductory comparisons, we will often quote the worst-case complexity unless stated otherwise.

## 9. Greedy algorithms versus optimal algorithms

So far, we have been interested in how efficiently an algorithm performs a task.

But there is another important question:

**Does an algorithm always find the best possible solution to a problem?**

An algorithm can be computationally efficient but still produce a solution that is not the best possible one.

This introduces the distinction between greedy algorithms and optimal solutions.

### 9.1 What is a greedy algorithm?

A greedy algorithm makes a sequence of decisions.

At each step, it chooses the option that appears to be the best choice at that particular moment.

It does not necessarily consider all possible combinations of decisions or look ahead to determine whether its current choice will lead to the best overall result.

The general approach is:

1. Look at the available choices.
2. Select the option that appears best locally.
3. Commit to that choice.
4. Repeat until the problem is solved.

The best decision at one particular step does not necessarily lead to the best overall solution.

### 9.2 Example: The knapsack problem

Imagine you are designing a robot that needs to collect valuable equipment.

The robot has a limited carrying capacity of 10 kg.

You have three objects available:

| Object | Weight | Value | Value per kg |
|---|---:|---:|---:|
| A | 6 kg | £12 | £2.00 |
| B | 5 kg | £9 | £1.80 |
| C | 5 kg | £9 | £1.80 |

The robot wants to maximise the total value of the objects it carries without exceeding its 10 kg capacity.

#### Approach 1: Greedy algorithm

A greedy approach might choose the object with the highest value per kilogram first.

1. Object A has the highest value per kilogram, so select it.
2. The robot now has 4 kg of capacity remaining.
3. Neither B nor C will fit.
4. The algorithm stops.

The greedy solution is object A:

- Total weight: 6 kg
- Total value: £12
- Unused capacity: 4 kg

#### Approach 2: Find the optimal solution

Instead of making the locally best decision, consider the possible combinations.

| Selected objects | Total weight | Total value |
|---|---:|---:|
| A | 6 kg | £12 |
| B | 5 kg | £9 |
| C | 5 kg | £9 |
| A + B | 11 kg | Not allowed |
| A + C | 11 kg | Not allowed |
| B + C | 10 kg | £18 |

The best solution is to select B and C.

- Total weight: 10 kg
- Total value: £18

The greedy algorithm found a valid solution, but it did not find the optimal solution.

### 9.3 Why is this important?

Greedy algorithms are often attractive because they are relatively simple and computationally efficient.

However, they do not always guarantee an optimal solution.

- **Greedy solution:** Makes the best available decision at each individual step.
- **Optimal solution:** Finds the best overall combination of decisions.

An algorithm that guarantees an optimal solution must account for the interactions between decisions, rather than simply making locally attractive choices.

### 9.4 When does a greedy algorithm work?

Greedy algorithms can produce optimal solutions for some problems.

For example, consider the **fractional knapsack problem**. In this variation, objects can be divided into smaller quantities.

The robot could take all of object A and then use its remaining capacity to take part of object B.

Choosing objects in order of value per kilogram produces an optimal solution for this version of the problem.

The important difference is that objects are divisible.

For the indivisible knapsack problem, the same greedy strategy does not guarantee an optimal solution.

## 10. Bringing everything together

We have introduced two related but distinct aspects of algorithm design.

**1. Computational efficiency**

How much time and memory does an algorithm require as the input size increases?

**2. Solution quality**

Does the algorithm produce a correct solution? Does it guarantee the best possible solution?

These are not the same thing.

An algorithm can be:

- Efficient and optimal.
- Efficient but not optimal.
- Computationally expensive but optimal.
- Computationally expensive and not optimal.

Consider the following examples:

| Algorithm | Main idea | Time complexity | What it guarantees |
|---|---|---:|---|
| Linear search | Check elements sequentially | \(O(n)\) | Finds a target if present |
| Binary search | Repeatedly halve the search space | \(O(\log n)\) | Finds a target if present in sorted data |
| Selection sort | Repeatedly select the smallest element | \(O(n^2)\) | Produces a sorted list |
| Insertion sort | Insert elements into a sorted section | \(O(n^2)\) worst case | Produces a sorted list |
| Merge sort | Divide and merge | \(O(n\log n)\) | Produces a sorted list |
| Greedy knapsack | Choose the highest value per unit weight | Depends on implementation | Not always optimal for indivisible objects |

Sorting algorithms are designed to produce a correctly sorted result. The difference between them is primarily how efficiently they achieve it.

For optimisation problems, the distinction between finding a valid solution and finding the best solution becomes particularly important.

## 12. Key takeaways

- **Algorithmic complexity** describes how computational resource requirements grow as the input size increases.
- **Big O notation** provides a way to describe the growth of computational work, independently of the particular computer being used.
- **Searching:** Linear search is \(O(n)\), while binary search is \(O(\log n)\), provided the data is sorted.
- **Sorting:** Simple sorting algorithms such as selection sort and insertion sort are generally \(O(n^2)\), while merge sort is \(O(n\log n)\).
- **Trade-offs:** Algorithms can make different compromises between computational time and memory requirements.
- **Greedy versus optimal:** Making the best decision at each step does not necessarily produce the best overall solution.
