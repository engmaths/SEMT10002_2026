# Algorithmic Complexity

Algorithmic complexity is a measure of the growth of the computational resources required by an algorithm as a function of its input size $n$. 

This week we are moving beyond simply asking *"Does this algorithm work?"* to asking *"Does this algorithm work efficiently, and will it continue to do so as the problem gets bigger?"*...

We are primarily interested in two measures: 
- **Time complexity**: how the number of *operations* grows as the input gets bigger
- **Space complexity**: how the amount of *memory* required grows as the input gets bigger

# Big O

We use "Big O" notation to describe complexity. 

Big O describes the order of the growth of the time or space required by an algorithm as a function of the input size $n$

Different algorithms may solve the same problem but require very different amounts of time and space

Importantly, in this context, time is *not* the growth rate of the of the execution time in seconds.

## $O(n)$

Algorithms of complexity $O(n)$:
- Need to process every item in the input dataset
- Processing one item takes approximately the same amount of time, regardless of the dataset size.

The maximum number of comparisons is *proportional to* the number of items $n$

This is described as **linear complexity** because the number of operations grows *linearly* with the input size.

The growth is therefore of order $n$ and the complexity is written

$$O(n)$$

### Example algorithm: Linear Search

Checks every item once until a target value is found
- Need to process every item in the input dataset
- Processing one item takes approximately the same amount of time, regardless of the dataset size.

````python
def linear_search(numbers, target):
    print('Input size', len(numbers))

    for i, number in enumerate(numbers):
        
        print(f'Comparison {i+1}: {number}=={target}')

        # COMPARISON
        if number == target: 
            return True

    return False
````

The number of comparisons is therefore equal to the number of items in the list

$$k=n$$


| Number of items $n$ | Max number of comparisons, $k$ |  
| :- | :- |  
| 1 | 1|  
| 5 | 5 |  
| 10 | 10 |  
| 100 | 100 |  
| 1,000 | 1,000 | 
| 10,000 | 10,000 | 





If there are **5 items** in the list to search, then the algorithm will perform a *maximum* of **5 comparisons**, $k$

````python
numbers = [4, 7, 2, 9, 5]
target = 5
linear_search(numbers, target)
````

Note that Big O notation is based on the *maximum* work done by the algorithm (operations/comparisons). 

Depending on the input conditions (e.g. the target is not the fianl item in this case), the actual number of operations may be less than the maximum. 

````python
numbers = [4, 7, 2, 9, 5]
target = 2
linear_search(numbers, target)
````

Suppose the list contains 1,000 elements.

| Situation | Number of comparisons |
|---|---:|
| Target is first | 1 |
| Target is in the middle | Approximately 500 |
| Target is last | 1,000 |
| Target is absent | 1,000 |

The best case is $O(1)$, but the worst case is $O(n)$.

Big O describes an upper bound on growth. 


### Example algorithm: Finding a maximum

Checks every item to find the largest in the series
- Need to process every item in the input dataset
- Processing one item takes approximately the same amount of time, regardless of the dataset size.

````python
def find_largest(numbers):

    print('Input size', len(numbers))
    
    largest = numbers[0]

    for i, number in enumerate(numbers):
        
        print(f'Comparison {i+1}: {number} > {largest}')

        # COMPARISON
        if number > largest:
            largest = number

    return largest
````

````python
numbers = [4, 7, 2, 9, 5]
print(find_largest(numbers), 'is largest number')
````

Again, the number of comparisons is equal to the number of items in the list

$$k=n$$


Notice, that becuase `largest` is initialised as `numbers[0]`, we can actually reduce the number of operations to

$$k = n-1$$

````python
def find_largest(numbers):

    print('Input size', len(numbers))
    
    largest = numbers[0]

    for i, number in enumerate(numbers[:-1]):
        
        print(f'Comparison {i+1}: {number} > {largest}')

        # COMPARISON
        if number > largest:
            largest = number

    return largest
````

````python
numbers = [4, 7, 2, 9, 5]
print(find_largest(numbers), 'is largest number')
````

#### Why is this still linear complexity?

As the input size increases the number of comparisons increases approximately in proportion to $n$.

Small order terms become negligable as the input size becomes large (the accuarcy of the approximation increases).

| Input size ($n$) | Total comparisons ($n-1$) |
|---:|---:|
| 1 | 0 |
| 2 | 1 |
| 5 | 4 |
| 10 | 9 |
| 100 | 99 |
| 1,000 | 999 |

To reflect this when calculating the complexity, **smaller-order** terms are ignored. 

Big O notation indicates the *order* of the growth of the work done by the algorithm as a funtion of the input size $n$, not the actual work done, so  **constant multipliers** are ignored. 

So an algorithm with number of comparisons $k = n-1$ has linear complexity $O(n)$

## $O(n^2)$

Algorithms of complexity $O(n^2)$:
- Need to perform up to $n$ operations (e.g. comparisons) per item
- Consequently, the total number of operations grows approximately with the square of the input size, $n^2$

This is described as **quadratic complexity** because the number of operations grows *quadratically* with the input size.

The growth is therefore of order $n^2$ and the complexity is written

$$O(n^2)$$

### Example algorithm: Finding repeat values

Each item is compared with every other item.

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/find_repeats_v1.png?raw=true" width="35%">

This is achieved using two loops

For each iteration of the outer loop $n$, there are $n$ iterations of the inner loop

Therefore there are $n^2$ comparisons made

````python
robot_names = ["Josh", "Hemma", "Martin", "Josh"]

def find_repeats_v1(items):

    repeat_values = []

    # N outer loop iterations
    for i, name_a in enumerate(items):

        # N inner loop iterations
        for j, name_b in enumerate(items):

            # COMPARISON
            if (name_a == name_b and i!=j): 
                repeat_values.append(name_a)

    return set(repeat_values)

find_repeats_v1(robot_names)
````

Now consider a slightly different way to approach this problem. 

We compare each item in the list, to the items to its right *only*. 

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/find_repeats_v2.png?raw=true" width="35%">

Each items has already been compared to the items to its left, so we don't need to repeat the operation

````python
robot_names = ["Josh", "Hemma", "Martin", "Josh"]

def find_repeats_v2(items):

    repeat_values = []

    # N outer loop iterations
    for i, name_a in enumerate(items[:-1]):

        # N-i inner loop iterations
        for name_b in items[i+1:]:
            if (name_a == name_b): 
                repeat_values.append(name_a)

    return set(repeat_values)

find_repeats_v2(robot_names)
````

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/triangular_numbers.png?raw=true" width="35%">

The total number of comparisons forms a triangular number series:

$k = (n-1) + (n-2) + ...  + 2 + 1$

If we add the series to the same series written in reverse, we get:

$2k = n + n + n + ... + n$

(There are $n-1$ terms each equal to $n$)

So we can simplify the number of comparisons to:

$$k = \frac{n(n-1)}{2} = \frac{n^2-n}{2}$$

So what is the complexity? 

Remeber that in Big O notation, **constant multipliers** and **smaller-order** terms are ignored 

So an algorithm with number of comparisons $k = \frac{n^2-n}{2}$ has linear complexity $O(n^2)$, as $n^2$ is the dominant term



So if we compare the two quadratic complexity algorithms implemented, their time complexity (order of growth with $n$) is the same category, $O(n^2)$ but `find_repeats_v2` requires approximately half as much work as `find_repeats_v1` for a given input size $n$

| $n$ | `find_repeats_v2`<br>$k=\frac{n^2-n}{2}$ |  `find_repeats_v1`<br>$k=n^2$ |
|:---|:---|:---|
| 2 | 1 | 4 |
| 5 | 10 | 25 |
| 10 | 45 | 100 |
| 20 | 190 | 400 |
| 50 | 1,225 | 2,500 |
| 100 | 4,950 | 10,000 |
| 500 | 124,750 | 250,000 |
| 1,000 | 499,500 | 1,000,000 |
| 10,000 | 49,995,000 | 100,000,000 |

| Algorithm / task | Example | Why is it $O(n^2)$? |
|---|---|---|
| Comparing items to all other items | Finding pairs of with matching characteristics | Each item is compared with every other item. |
| Duplicate detection (nested loops) | Checking whether a list contains duplicate values | Each item may be compared against every other item. |
| Selection sort | Finding the smallest remaining item and moving it into position | Repeatedly searches through the remaining unsorted items. |
| Insertion sort | Inserting each new item into its correct position in a sorted section | Repeatedly searches through the sorted items.  |
| Bubble sort | Sorting a list by repeatedly comparing adjacent elements | Repeatedly traverses the list, making comparisons between elements. |

## $O(n^x)$

### Example algorithm: Generate all $x$-digit combinations

Nested loops can tell us something about the complexity of an algorithm

The time complexity increases exponentially with the number of nested loops iterating over $n$ items

- One loop over $n$ items → $O(n)$.
- Two nested loops, each iterating over $n$ items → $O(n^2)$.
- Three nested loops, each iterating over $n$ items → $O(n^3)$.



Find all combinations of $x=$ 3 digits, where each digit has $n=$ 2 possible values 

How does it effect the number of operations:
- if $x$ increases?
- if $n$ increases?

````python
combinations = []

input = [0,1]

for a in input:
    for b in input:
        for c in input:
            combo = [a, b, c]
            combinations.append(combo)

print('Number of operations =', len(combinations))

for i in combinations:
    print(i)
````

## $O(\log n)$

Algorithms of complexity $O(\log n)$:
- **Reduce the problem size at each step:** <br>Each operation removes a substantial proportion of the remaining possibilities, often half.
- **Do not need to examine every item:** <br>Instead of processing the entire dataset, they progressively narrow down where the answer could be.
- **The number of steps grows slowly:** <br>Doubling the input size only adds approximately one extra step (logarithmic relationship)

### Example algorithm: Binary search

Now consider binary search. This algorithm works on a sorted list.

Rather than checking every item, every comparison divides the remaining search area by a factor of 2.

<img src="https://github.com/engmaths/SEMT10002_2026/blob/main/media/week_9/binary_search.png?raw=true" width="35%">

````python
def binary_search(numbers, target):

    # Upper and lower bound of range to check
    low = 0
    high = len(numbers)

    # When the upper and lower bound overlap, we have checked the whole list
    while low < high:

        # Find the middle index
        middle = (low + high) // 2

        # ----------------------------
        # COMPARISONS
        if numbers[middle] == target:   
            return True
        elif numbers[middle] < target:
            low = middle + 1    # Increase lower bound 
        else:
            high = middle - 1   # Reduce upper bound

        print(low, high)
        # -----------------------------

    return False

````

Consider a sorted list of 11 numbers

````python
numbers = [2, 5, 8, 12, 16, 23, 38, 45, 56, 72, 91]

binary_search(numbers, 41)
````

Each comparison approximately halves the number of remaining items to search.

What is the maximum number of comparisons required for $n$ items? <br>
(i.e. how many times do we need to half the number of items to be left with 1 item?)

| Items, $n$ | Comparisons, $k$ |
|:--|:--|
| 1 | 0 |

| 4 | 2 |
| 8 | 3 |

Notice that $n$ increases exponentially with $k$:

$n = 2^k$

To find the inverse of this function, we take the logarithm base 2 of both sides:

$k = \log_2(n)$


This algorithm therefore is described as having **logarithmic complexity** because the number of operations grows *logarithmically* with the input size.

The growth of the algorithm is therefore of order $\log (n)$ and the complexity is written

$$O(\log n)$$




Even with a very large number of items the computational work remains relatively small.

We round up to find the maximum number of operations as this must be an integer 



So we can see, that for an already sorted list, binary search is far more effient than linear search.

## $O(1)$

Algorithms of constant complexity $O(1)$:
- **A fixed number of operations:** <br>They perform approximately the same number of operations regardless of the size of the input dataset.
- **No need to traverse the dataset:** <br>They do not need to examine every item or search through the input.
- **Direct access to information:** <br>They often use an index, key or previously stored information to access the required data.
- **Input size does not affect the number of operations:** <br>Doubling the dataset size does not increase the number of operations.

## How is algorithmic complexity calculated? : **Space complexity**


Algorithmic **space** complexity describes how the **amount of memory** required by an algorithm changes as the input size $n$ increases.

Space complexity usually refers to *additional memory* used by the algorithm, rather than the memory already occupied by the input.
 


Just as with time complexity, Big O is used to describe space complexity:

| Space complexity | Example | Why? |
|---|---|---|
| $O(1)$ | Finding the largest number in a list | Only a fixed number of additional variables are needed, regardless of the list size. |
| $O(n)$ | Creating a copy of a list | The new list requires space proportional to the number of input items. |
| $O(n^2)$ | Creating an $n \times n$ matrix | The matrix contains $n^2$ elements. |
| $O(\log n)$ | Storing the remaining unseacrhed items at each step of a binary search | Each iteration halves the remaining items. |

Space complexity becomes a bit difficult to show through elementary examples as it requires a knowldge of underlying mechanisms of computer memory that is outside the scope of this unit. 

But we can get a general understanding through some simple code examples:

## $O(1)$
### Constant space

Our function from earlier that finds the largest number in a list.

Only one additional variable (largest) is needed, regardless of how many numbers are in the list.

````python
def find_largest(numbers):
    
    largest = numbers[0]

    for number in numbers[:-1]:

        if number > largest:
            largest = number

    return largest

numbers = [4, 7, 2, 9, 5]
find_largest(numbers)
````

## $O(n)$
### Linear space

Creating a copy of a list.

Every additional input item requires another element in the new list.

````python
def copy_list(numbers):
    new_list = []

    for number in numbers:
        new_list.append(number)

    return new_list

numbers = [4, 7, 2, 9, 5]
copy_list(numbers)
````

## $O(n)$
### Linear space (dictionary)

Creating a dictionary containing the square of each number.

````python
def create_squares(numbers):
    squares = {}

    for number in numbers:
        squares[number] = number ** 2

    return squares

numbers = [4, 7, 2, 9, 5]
create_squares(numbers)
````

## $O(n^2)$
### Quadratic space 

Creating a multiplication table.

For an input of size $n$, the function creates an $n \times n$ matrix.

(Note that the nested loops in the multiplication table are not what make its space complexity $O(n^2)$. It is the fact that we store $n^2$ elements in memory.)



````python
def multiplication_table(n):
    table = []

    for i in range(1, n+1):
        row = []

        for j in range(1, n+1):
            row.append(i * j)

        table.append(row)

    return table

multiplication_table(5)
````

### Exercise 2

Selection sort:
- iterates through a list
- repeatedly finds the smallest element in the unsorted part of a list
- swaps it with element at the index of the loop counter
- continues until the entire list is sorted.

Analyse the complexity of this algorithm:

- How many times is the comparison applied for a given input size?
- How does this scale with the input size? 
- Which of the examples we have studeied today does this look most similar to? 
- What do you think the Big O complexity of the algorithm is? 


````python
def selection_sort(numbers):
    """
    Sorts a list of values in ascending order in-place
    """

    # Iterate through the list
    for i in range(len(numbers)):

        # Initialise index of the miniumum value as the loop counter
        min_index = i

        # Update index of the miniumum value as smallest value in unsorted part
        for j in range(i + 1, len(numbers)):
            if numbers[j] < numbers[min_index]:
                min_index = j
        
        # Swap smallest value and value at index of loop counter
        numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

numbers = [1,2,5,7,2,5,6]
selection_sort(numbers)
print(numbers)
````
