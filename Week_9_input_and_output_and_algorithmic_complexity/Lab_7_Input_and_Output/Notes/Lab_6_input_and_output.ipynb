### What is a file? 

A file is a set of bytes (8 bits) used to store data. 

What the data represents depends on the file type which is represented by the file extension. 

Examples of file types and file extensions:
- unformatted text (.txt, .dat)
- formatted text (.docx)
- spreadsheet/tabulated data (.xlsx, .csv)
- image (.png, .img)

(there are hundereds more)

The file type determines what these bytes represent. 

__Text files__: Human-readable data. <br>Bytes represent plain text characters <br>e.g. .py, .csv, .json, .txt

__Binary files__: Data that is not intended to be human-readable. <br>Bytes do not represent plain text characters, but other information about the file. <br>e.g. executable programs (.exe, .bin), images (.jpg, .png, .gif), audio (.mp3, .wav), video (.mp4, .avi), compressed files (.zip). 

**For readbility, we will be working with data stored in *text file* formats for the rest of the unit.**

### Using imported data within a program

File objects are *iterable*: each item is a new line of the file

````python
with open('README.txt') as file:
    for value in file:
        print('Line:', value)
````

File objects are not *subscriptable* (we can't access an individual element using an index)

````python
with open('README.txt') as file:
    print(file[0])
````

Once the indented block after the `with` statement ends, the file is closed.

The variable that points to the file object goes out of scope.


````python
with open('README.txt') as file:
    print(file.read())

# print(file.read())
````

We can *cast* the file object as different object type that makes it easier to manipulate the data within the computer program

By casting the file object as a list, the data:
- is iterable
- is subscriptable
- can be accessed (remains in scope) once the file is closed

Each element of the list is a new line of the file

````python
with open('README.txt') as file:
    data = list(file)

    # Iterable
    print("Printing all lines of the file...")
    for value in data:
        print(value)

    # Subsciptable
    print("Printing first line of the file...")
    print(data[0])

# Remains in scope after the file is closed
print("Printing second line of the file...")
print(data[1])
````

We can transform this single string to separate items:
1. Convert file object to list
2. Convert single string to list of values with commas removed using `split`
3. Remove the new line character from any strings in the list using `replace` (remember the first example in the notes!)

````python
with open('temperature.csv') as file:
    data = list(file)         
# Remove any new line characters 
    for line in data:     

        # Split into list of values 
        values = line.split(',')  

        # Remove any new line characters 
        values = [v.replace('\n', '') for v in values] 

        print(values)
````

Notice that the items of each list are still string data. 

If we want to use numerical values as numbers within our program we need one more step to convert the values to numerical data

````python
with open('temperature.csv') as file:
    file = list(file)

    # Second row contains numerical data
    values = file[1]

    # Split into list of values 
    values = values.split(',')

    # Remove any new line characters 
    values = [v.replace('\n', '') for v in values] 

    # Convert to numerical data
    numbers = [float(v) for v in values]
    
    print(numbers)
````

### Variations in filepath expression

#### File paths on different operating systems 

The syntax for the file path is different on Windows and Mac/Linux

**Mac/Linux**: Directories are separated by a forward slash character `/` <br>(This syntax is used in the examples above)
**Windows**: Directories are separated by a backslash character `\` <br>(Substitute into the examples above if working on Windows

#### Local and Global file path

These examples show the:<br>
**Local (relative)** file path: The path to a file relative to the current *working directory* (the directory where the program is being run) 

The file path can also be expressed as the:<br>
**Global (absolute)** file path: The path to a file from the **root directory** of the file system. The root directory is the top-most directory of the file system.

##### Linux/Mac

Example global path: '<span style="color:blue">/home/YourUsername/Documents/</span><span style="color:red">myfile</span><span style="color:green">.txt</span>'
<br><span style="color:black">

There is a single root directory for the entire file system, denoted by a forward slash `/`

##### Windows

Example global path: 
'<span style="color:blue">C:\Users\YourUsername\Documents\ </span><span style="color:red">myfile</span><span style="color:green">.txt</span>'
<br><span style="color:black">

Each drive has its own root directory:
- `C:\` is the root directory of the C: drive.
- `D:\` is the root directory of the D: drive.

#### Reading a CSV file

We can use the `reader` function to read a csv file

`reader` automatically converts a row of data seperated by commas into a list, and avoids the need to break a single string up into individual items. 

Input arguments:
- The csv file to read
- (Optional) formatting parameters

Returns:
- A reader object which, like a file object, is *iterable* but not *subscriptable*.
- We can convert the reader object to a list if we want to access/manipulate the values

Additional information:
- *If csvfile is a file object, it should be opened with newline=''.*
- e.g. `with open('my_file.csv', newline='') as file:`
- This allows the CSV module to correctly handle line endings, including newlines inside quoted fields. 


````python
with open('temperature.csv', newline='') as file:
    
    data = csv.reader(file)
    
    for value in data:
        print(value)
````

#### Writing a csv file

We can use the `writer` function to read a csv file

Input arguments:
- The csv file to read
- (Optional) formatting parameters (e.g. the mode)

Returns:
- A writer object
 

````python
import csv

with open('output_data.csv', 'w', newline = '') as file:
    
    writer = csv.writer(file)

    '''
    Writes input argumnent to the file we opened, must be an iterable (e.g. list)
    '''
    writer.writerow([1,2,3])    


    '''
    Writes input argumnent to the file we opened
    Must be an iterable of iterables (e.g. nested list)
    '''
    writer.writerows([[1,2,3],[4,5,6]])
````

### The sys module


To access the command line arguments within a program we can use the biult in sys module. 

The command-line arguments are stored as a list of **strings** assigned to variable sys.argv

- `sys.argv[0]` → the script name

- `sys.argv[1]`, `sys.argv[2]`, etc. → the arguments you pass

### Example: Numerical command line arguments
**add_numbers.py**

***


### Flags

A flag is a special command-line argument that switches a particular feature on or off.

Flags allow users to control optional features of a program when they run it, without needing to change the Python code.



### Example: Flags

**robot.py**
***
```Python
import sys

if "--debug" in sys.argv:
    print("Debugging mode enabled")
else:
    print("Normal mode")
```
***

**In a terminal**
```python 
python robot.py --debug
```

### Naming conventions

| Type | Example | Purpose |
|---|---|---|
| Value option | `--speed 20` | Sets a value |
| Boolean flag | `--debug` | Switches a feature on |
| Boolean flag | `--no-display` | Switches a feature off |

**Double hyphen (--)**: Used for long, descriptive argument names.
```Python
python robot.py --speed 20 --debug
```

**Single hyphen (-)**: Used for short, usually single-letter abbreviations.
```Python
python robot.py --s 20 --d
```
