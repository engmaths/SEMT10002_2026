## Exercise 3 - Fixing a factorial function

The function below is *supposed* to calculate the factorial
$n! = 1 \times 2 \times 3 \times \ldots \times n$, so `factorial(4)` should give
$1\times2\times3\times4 = 24$.

It doesn't work — some of the tests below fail. **Don't rewrite it from scratch.** Instead:
1. Run it and look at what it actually returns.
2. Work out *why* — trace it by hand for `n = 3`, writing down `result` after each iteration.
3. Make the smallest change you can to fix it.

````python
def factorial(n):
    result = 1
    for ii in range(n):
        result = result * ii
    return result


# === Do not edit below this line ===
if abs(factorial(0) - 1) < 1e-5:
    print('factorial(0) passed')
else:
    print('factorial(0) FAILED: got', factorial(0), 'expected 1')

if abs(factorial(1) - 1) < 1e-5:
    print('factorial(1) passed')
else:
    print('factorial(1) FAILED: got', factorial(1), 'expected 1')

if abs(factorial(4) - 24) < 1e-5:
    print('factorial(4) passed')
else:
    print('factorial(4) FAILED: got', factorial(4), 'expected 24')

if abs(factorial(6) - 720) < 1e-5:
    print('factorial(6) passed')
else:
    print('factorial(6) FAILED: got', factorial(6), 'expected 720')
````

 This week, we're going to use a loop to calculate the final position of the robot after it moves for a certain amount of time. The basic idea is that if we know the speed, we can integrate to calculate the new position. In general, if we have something at position $x$, moving with speed $v_x$ for time $dt$, then the new position $x'$ is given by:

 $$ x' = x + (dt * v_x) $$

 which is nothing more than distance = speed x time when we think about it. 
 
 The complication is that our robot has three state variables and the new positions depend on the value of these state variables as well as the wheel speeds. If the robot (at x=0, y=0) has an angle $\theta = 0$, then a linear speed of 1 m / s applied for 1 second will lead to a new position of (x=1, y=0). If the robot has instead an angle of $\theta = 90$, then we'll get a position of (x=0, y=1). This means we *cannot* just do:

 $$
 x' = x + (\cos(\theta) * T * v) \\
 y' = y + (\sin(\theta) * T * v) \\
 \theta ' = \theta + (T * \omega)
 $$

 where T is the total duration of the movement. Doing so assumes that the value of $\theta$ stays the same for the entire movement -- which we can see from the third equation it doesn't. 

 Instead, we have to break the total movement up into a sequence of small steps -- say each consisting of a period of time equal to 100 milliseconds. For short movements like this, the assumption that $\theta$ doesn't change is still incorrect -- but the error it introduces is much smaller than if we just apply a single update.
