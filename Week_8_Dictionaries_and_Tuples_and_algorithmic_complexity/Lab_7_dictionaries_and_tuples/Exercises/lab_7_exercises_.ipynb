The breadth-first and depth-first search algorithms can be extended to graphs containing cycles, like the FSM or the road network we saw earlier. 

Without keeping track of visited nodes, a search, as we have implemented it so far, could loop forever.

Updated function implementations shown below, each includes an additional operation to record nodes visited, and avoid adding already visited nodes to the queue (BFS) or stack (DFS) a second time  

````python
def bfs(graph, start):
    """
    Uses the breadth-first search algorithm to print the 
    items of a graph/network level by level
    """

    queue = [start]

    # [UPDATE] Mark each node as visited when it is added to the queue
    visited = [start]

    # Keep going while there are still nodes waiting to be visited
    while queue:
        print('queue:', queue)

        # Remove the FIRST node added to the queue
        node = queue.pop(0)
        print('node visited:', node)

        # Add all the node's children to the queue [UPDATE] if they haven't been visited
        for neighbour in graph[node]:
            if neighbour not in visited:
                queue.append(neighbour)
                visited.append(neighbour)

````

````python
def dfs(graph, start):
    """
    Uses the depth-first search algorithm to print the items of a graph/netork branch by branch
    """
    # Start with a stack containing the node we want to visit first
    stack = [start]

    # [UPDATE] Mark each node as visited when it is added to the stack
    visited = [start]

    # Keep going while there are still nodes waiting to be visited
    while stack:
        print('stack:',end="\n")
        for i in reversed(stack):
            print(i)
        print()

        # Remove the LAST node added to the stack
        node = stack.pop()
        print('node visited:', node)
        
        # Add all the node's children to the stack
        for neighbour in graph[node]:
            if neighbour not in visited:
                stack.append(neighbour)
                visited.append(neighbour)
            
        # # To traverse in the same order as breadth first search
        # for neighbour in reversed(graph[node]):
        #     if neighbour not in visited:
        #         stack.append(neighbour)
        #         visited.append(neighbour)

````

We can extend this representation by associating an event with each transition. 

When an event is given, the condition and the robot's current state determines the next state that the robot transitions to. 

For example, the robot transitions from IDLE to MOVING when the start button is pressed, or from MOVING to AVOIDING_OBSTACLE when an obstacle is detected.

````python
robot = {
    "IDLE": {
        "start_button": "MOVING",
        "battery_low": "CHARGING"
    },

    "MOVING": {
        "stop_button": "IDLE",
        "obstacle_detected": "AVOIDING_OBSTACLE",
        "battery_low": "RETURNING_TO_CHARGE"
    },

    "AVOIDING_OBSTACLE": {
        "path_clear": "MOVING",
        "stop_button": "IDLE"
    },

    "RETURNING_TO_CHARGE": {
        "charger_found": "CHARGING",
        "obstacle_detected": "AVOIDING_OBSTACLE"
    },

    "CHARGING": {
        "battery_full": "IDLE",
        "start_button": "MOVING"
    }
}


def transition(robot, current_state, event):
    if event in robot[current_state]:
        return robot[current_state][event]

    # No transition defined for this event
    return current_state


state = "MOVING"

event = "obstacle_detected"
state = transition(robot, state, event)
print(state)  # AVOIDING_OBSTACLE

event = "path_clear"
state = transition(robot, state, event)
print(state)  # MOVING
````
