Some code to compute and display the area of three circles 


````python
# Solution without functions (10 lines)

pi = 3.14159                        # Define a value for pi, setting the nunber of decimal places

radius1 = 3                         # Define the radius of circle 1   
diameter1 = 6                  
area1 = pi * (diameter1/2) ** 2           # Compute the area of circle 1 
print('Area of circle:', area1)     # Display the area of circle 1

radius2 = 5  
diameter2 = 10                       
area2 = pi * (diameter2/2) ** 2           # Compute the area of circle 1         
print('Area of circle:', area2)

radius3 = 7                         
area3 = pi * radius3 ** 2           
print('Area of circle:', area3)
````

````python
# Solution with functions (6 lines)

from math import pi

def circle_area(diameter):                # A function to compute and display the area of a circle using its radius
    area = pi * (diameter/2) ** 2
    print('Area of circle:', area)      

circle_area(6)
circle_area(5)
circle_area(7)           # Compute the circle area
````

## What *actually* happens when we call a function?

We saw in last week's seminar that a computer can store data in its *memory*. This data represents the *state* of the computer -- given the state and one instruction, we know everything we need to know to calculate the next state.  We also saw how a CPU uses the *program counter* to keep track of which instruction should be executed next. This represents our mental model of what a Python program actually does. How we can extend this to include functions?

Let's start with a simple example. We'll run through the code without using functions first, then see what's different when we re-write our code using a function.

````python
a = 1
b = 2
sum1 = a + b
print(sum1)
````

Pretty simple right -- we just go through the instructions one by one. Now let's see what happens when we use a function to do the same thing for us.

````python
def add(a, b):
    return a+b

x = 1
y = 2
sum1 = add(x, y)
print(sum1)
````

OK, so what happens here when we get to the first instruction? 

When Python *executes* a ```def``` statement, it does *not* run the function body. Instead, it creates a box in memory (instructions are just another type of data!). We saw in week 1 that each piece of data stored in memory must have a name, a value and a type. Here, the name is simply *add* (the name of the function), the data type is *function* and the value is the *function body* -- i.e the instructions that make up the function. The program counter will skip over any code that is contained in the function body to the next instruction. So after executing one instruction (```def add(a, b))```), our state will be as below:

<img src="Slide6.png" width="45%">
<img src="Slide7.png" width="45%">

Running the next few instructions, we get the following sequence of states:

<img src="Slide8.png" width="45%">
<img src="Slide9.png" width="45%">

The next line to execute is line 6 -- when we actually call the function. So what happens here? We can think of a function call as consisting of six steps:

1. Remember the *return address* -- on line 6, we're assigning the variable *sum1* with whatever add outputs. 
2. Create some new variables to store the function inputs (*x*, *y*). Note that these are **not** the same as the variables we created on lines 4 and 5. 
3. Copy the values of the input (a, b) into the new variables.
4. Set the program counter to the first line of the function body.
5. Run the body until we hit return.
6. Set the program counter back to the line we were on before the function call and store any output values.

Let's see each of these in sequence. First, we create the variable *sum*. We don't have a value for it yet, but we'll need it to store whatever the function outputs.

<img src="Slide10.png" width="45%">

Next, we make some new variables to store the function inputs. As these variables are associated only with the function *add*, we put them inside their own *scope*. This means they can *only* be accessed from within the function *add* **and** will be deleted once the function returns.

<img src="Slide11.png" width="45%">

Next, we copy the values of the inputs to our new variables. 

<img src="Slide12.png" width="45%">

We're now ready to actually execute the function -- so we set the program counter to 2 as that's the first line of the function body.  

<img src="Slide13.png" width="45%">

Upon returning, we store the output of the function (3) in the variables *sum1* that we remembered right at the start of this process. As the function has ended, we also remove the variables associated with the function from memory and set the program counter to the line after the function call.

## Nested Function calls

We've seen what happens when we call a function -- briefly, the CPU will create some new memory to store return address, arguments and any local variables. The CPU will then jump to start of the function body (by setting the program counter), proceeding step by step, until it hits a *return* statement. At this point, the memory associated with the function will be deleted, the return value of the function (if there is one) will be stored, and the program counter will be set back to what it was before the function call. The memory associated with a function -- the return address, arguments and local variables are known as the *stack frame*. 

What happens when we call a function from within another function? Let's look at an example!

````python
def square(n):
    return n**2

def sum_of_squares(a, b):
    return square(a) + square(b)

a = 1
b = 2

print(sum_of_squares(a, b))
````

As with the previous example, when we meet the ```def``` on line 1, Python stores the *square* function in memory and updates the program counter to line 4 - everything in the function body is skipped for now. 

<img src="Slide15.png" width="45%">
<img src="Slide16.png" width="45%">

At line 4, we get another ```def``` and so the same thing happens again -- we store the function header in memory and skip to line 7.

<img src="Slide17.png" width="45%">

On lines 7 and 8, we store our two global variables in memory.

<img src="Slide18.png" width="45%"><img src="Slide19.png" width="45%">

When we get to line 10, we call *sum_of_squares*. This creates the stack frame for the sum of squares function.

<img src="Slide20.png" width="45%">

Next we execute the body of sum_of_squares on line 5. This line calls the function *square*, so we start by making the stack frame for the square function. Note the scope of this frame -- it is inside the stack frame for the sum_of_squares function.

<img src="Slide21.png" width="45%">

When square returns, we go back to line 5 and delete the stack frame associated with the call to square.

<img src="Slide20.png" width="45%">

We then call the *square* function again, which creates a new stack frame.

<img src="Slide22.png" width="45%">

We then run line 2 again (with n=2) and return. Upon returning, we delete the square stack frame and jump back to line 5. This causes the sum_of_squares function to return, leading to the deletion of the sum_of_squares stack frame and the program counter being set back to line 10. At this point, our code prints the output from sum_of_squares.
