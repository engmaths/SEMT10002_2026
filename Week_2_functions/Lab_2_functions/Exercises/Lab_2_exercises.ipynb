Over this course, we're going to build a movement simulator for a simple 2 wheeled robot. 

<img src="https://github.com/engmaths/SEMT10002_2025/blob/main/media/week_1/romi.jpg?raw=true" width="25%" />

Our robot moves by spinning its two wheels independently -- this is called a *differential drive*. Here's the small amount of theory you need:

+ A wheel of radius $r$ spinning at angular speed $\omega$ (in radians per second) drives the robot forward at speed $v = r\omega$ at that wheel.
+ When both wheels spin at the same speed, the robot drives in a straight line. When they differ, it turns.
+ The robot's overall **forward speed** is the *average* of its two wheel speeds.
+ The robot's **turn rate** (radians per second) is $(v_{right} - v_{left}) / W$, where $W$ is the wheel separation.

## Exercise 5 - The Register Machine

### Part 1

In last week's seminar, we saw that a computer program is a set of instructions that a CPU can execute. Let's make this idea a bit more concrete by building a simulation of a very simple CPU, known as a register machine. We'll then program it in a form of "machine code". 

The CPU we'll simulate is deliberately simple. It has:

- Three registers (i.e blocks of memory), each of which can hold a single integer.
- A program (i.e a list of instructions and data).
- A program counter (pc) that tracks which line should be executed next. 

For now, our machine only has five instructions (we may add more later):

| Instruction   | What it does                                                    |
|---------------|-----------------------------------------------------------------|
| `('inc', r)`  | add 1 to register `r`, then move to the next line               |
| `('dec', r)`  | subtract 1 from register `r`, then move to the next line        |
| `('jmp', n)`  | jump to line `n` (i.e set the program counter to n)             |
| `('jz', r, n)`| if register `r` is 0, jump to line `n`; otherwise next line     |
| `('halt',)`   | stop the machine  

A 'program' for our computer will be a list, with each element containing one instruction and optionally some data.  For example, the program for add is:

```
P_add = [('jz', 1, 4),  # if r1 == 0 then jump to line 4 (finished)
         ('inc', 0),    # r0 += 1
         ('dec', 1),    # r1 -= 1 
         ('jmp', 0),    # go back to the top
         ('halt', )       #stop 
         ] 
```

The first element is the tuple ('jz', 1, 4). Here, 'jz' is the instruction and 1 and 4 are the data. Our simulator will read one instruction at a time (fetch), extract the instruction (decode) and then apply that instruction to the data (execute).  

Below is a half-built simulation of this machine. We've given you the fetch-decode-execute cycle and the 'halt' and 'inc' instructions. Your job is to complete the code so that we can run the program 'add' which computes r0 = r0 + r1. 

````python
#Simulate a mininal register machine 
def run(program, registers, max_steps=10000):

    #program counter - which line we're on
    pc = 0
    #number of lines executed                                 
    steps = 0

    while pc < len(program):
        steps += 1
        if steps > max_steps:
            raise RuntimeError('step limit exceeded — is your program looping forever?')

        #fetch the next instruction
        instruction = program[pc]
        #decode the instruction
        op = instruction[0]       

        if op == 'halt':
            break

        elif op == 'inc':
            #get register
            register = instruction[1]
            #add one to the target register
            registers[register] = registers[register] + 1
            #increment the program counter
            pc += 1

        elif op == 'dec':
            # subtract 1 from register instruction[1], then move to the next line
            pass   # Your code here

        elif op == 'jmp':
            # set the program counter to line instruction[1]
            pass   # Your code here

        elif op == 'jz':
            # if register instruction[1] is 0, jump to line instruction[2];
            # otherwise move to the next line
            pass   # Your code here

        else:
            raise ValueError('unknown instruction: ' + str(op))

    return registers

'''Add register 1 to register 0, storing the result in register 0. 
Note that the program consists of a sequence of tuples of the form (instruction, data1, data2, ..., dataN)
'''
P_add = [('jz', 1, 4),  # if r1 == 0 then jump to line 4 (finished)
         ('inc', 0),    # r0 += 1
         ('dec', 1),    # r1 -= 1 
         ('jmp', 0),    # go back to the top
         ('halt', )       #stop 
         ] 
registers = [5, 3, 0]

print(run(P_add, registers))
````
