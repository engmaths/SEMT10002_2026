# Functions

In seminar 1, we said that designing an algorithm means breaking a big task down into a sequence of smaller steps. Each step may itself be composed of even smaller steps -- and it's helpful to be able to group some instructions together and think of them as completing a subtask. A function is how you implement that idea in Python — it's a named, reusable block of code that performs one specific task and can be combined to solve bigger problems.

If we think back to our making a cup of tea example, we saw that the "add milk" step actually involves a number of smaller steps -- we can sort of think of this as a "function" for adding milk. 

You've already been using functions since week 1. Every time you write print(...), type(...), or float(...), you're calling a function. Importantly, you don't need to know how print works internally to use it. That separation between what a function does and how it does it is known as *abstraction* and is one of the most powerful ideas in programming: it lets you build up complex programs from small, well-defined pieces, each doing exactly one job.

## Using functions

Python has a number of built-in functions (e.g. print, type) that we've already used. These functions take something as input (either a value or a variable) and then do something with it. Formally, we refer to using a function as *calling* a function. To call a function, we write the name of the function, followed by round brackets '()' containing any inputs.

````python
message = "Hello, World!"
print(message)
````

You can find a list of all the built-in functions <a href=https://docs.python.org/3/library/functions.html>here</a>. As you can see it's quite minimal -- what happens if we need a function that's not in this list? We could write it ourselves (more on that later), but generally it's easier if we don't. Fortunately for us, many useful functions have been collected into libraries (again, more on these later) that we can *import* into our code and use. As with the built-in functions, we don't need to know exactly *how* these work-- just what they do. 

For example, if we wanted to calculate the sine of an angle, we could import the sin function from the math library and call it as below.

````python
from math import sin
x = 0.63
y = sin(x)
print(y)
````

We saw in seminar 1 that the CPU in our computer knows how to do basic operations: add, multiply, compare, etc. We then said that designing an algorithm that runs on a computer means iteratively breaking a task down until all of the steps remaining are ones that our CPU can execute. In practice, if every piece of code we wrote required us to keep breaking a task down this far, we'd not get much done! Instead, rather than breaking a task down into these fundamental operations, we can break it down into functions -- and as long as the functions have been written (by us or someone else), we're done.



### Interfaces

Look again at the example `y=sin(x)`.
* `x` is an _argument_, meaning that the function code is passed `x` for use in its calculations.  You can think of these as inputs to the function.
* `y` is the variable where the _return value_ is stored.  You can think of this value as the output.

That's the basic concept of the function _interface_: we give it arguments, and it returns values. 

Functions do not always return values, as their purpose might be to take some other action, e.g. `print`. Note that we can still use a function with no return value to assign a value to a variable (```output = print(message)```), but this will simply set the variable to be a "NoneType" -- Python's way of representing "nothing here". 

````python
function_input = "Print has an (optional) input but no output"
output = print(function_input)
print('Output of print is of type:', type(output))
print('Output has value:', output)
````

Functions also might not need arguments, as they might not need to use our data, e.g. getting the time.

````python
from time import time
t = time()
print(t)
````

Functions can have multiple arguments and return values.

````python
from math import atan2, degrees
angle = atan2(-1,-1)
print(degrees(angle))
````

We can also use the output of one function as the input of another function.

````python
from math import sin, atan2
value = sin(atan2(-1, 1))
print(value)
````

### Dot Notation

Some functions are attached to particular objects -- for example, if we have a string

````python
my_string = "hello"
print(my_string)
````

We can turn the letters in this string from lower case to upper case by writing ```my_string.upper()```. Here, we are calling the function (technically *method*) upper with the variable my_string as an input. i.e it's a bit like writing ```upper(my_string)```. We'll cover the difference between functions and methods more fully later, so don't worry if you don't totally get how this works. Often, dot notation will be used to chain together a sequence of operations into a single line.

````python
from math import sin, atan2 

# Chaining function calls with standard syntax
value = sin(atan2(-1, 1))

# Chaining functions calls with dot notation
result = "Hello world   ".strip().lower().replace("hello", "hi")
print(result)
````

Here, we in sequence strip spaces from the string (.strip()), convert it to lower case (.lower()) and then replace the substring "hello" with the string "hi". Note that .replace has three inputs (the output from the previous step, "hello", and "hi").

## Writing functions

Alas, not all the functions that we'll need for our code have been written for us. In general, we'll need to use a combination of built-in, library and user-defined functions to write our code.  Before showing examples, let's discuss why.

### Why functions?

Using functions brings about two key benefits:

1. Enables *abstraction*.
2. Reduce code repetition.

Consider the following code for finding the distance between two points.

````python
from math import sqrt

xa = 1
ya = 1
xb = 2
yb = 2

distance_ab = sqrt((xb-xa)**2 + (yb-ya)**2)

print('Distance from (',xa,',',ya,') to (',xb,',',yb,') is',distance_ab)
````

If we want to do the same job twice, it's not too hard to copy and paste:

````python
from math import sqrt

xa = 1
ya = 1
xb = 2
yb = 2

distance_ab = sqrt((xb-xa)**2 + (yb-ya)**2)
print('Distance from (',xa,',',ya,') to (',xb,',',yb,') is',distance_ab)

xc = 1
yc = 3

distance_ac = sqrt((xc-xa)**2 + (yb-ya)**2)
print('Distance from (',xa,',',ya,') to (',xc,',',yc,') is',distance_ac)
````

> Is this right?  What's gone wrong?

Generally speaking, do not copy and paste code.  It invites errors when you tweak it.  Better, package this job using a function, and call the function twice.

````python
from math import sqrt

def distance(x1,y1,x2,y2):
    d12 = sqrt((x1-x2)**2 + (y1-y2)**2)
    return d12

xa = 1
ya = 1
xb = 2
yb = 2
distance_ab = distance(xa,ya,xb,yb)
print('Distance from (',xa,',',ya,') to (',xb,',',yb,') is',distance_ac)
xc = 1
yc = 3
distance_ac = distance(xa,ya,xc,yc)
print('Distance from (',xa,',',ya,') to (',xc,',',yc,') is',distance_ac)
````

Better - the calculation is coded once, and we re-use it by calling the function with different arguments. This will massively reduce the number of bugs in our code, simplify editing and make our code easier to read and reason about.

It's also let us separate *what* the function does from *how* it does it. When I'm reading line 11 or 15, I can focus on the 'what', only concerning myself with the *how* (lines 3-5) if necessary. 

### Function syntax

In the example below, the keyword `def` marks the start of the function, and then we give it a name ('distance'), and specify some arguments (x1, y1, x2, y2).

````python
from math import sqrt

def distance(x1,y1,x2,y2):
    d12 = sqrt((x1-x2)**2 + (y1-y2)**2)
    return d12
````

The *body* of the function is the indented bit, and it will terminate when it gets to a `return` statement, giving the value of the expression after `return` as the return value of the function. The body of the function is where we write the instructions that we want the function to perform. Any code after the return statement of a function will not be executed as the function has terminated by the point.

````python
from math import sqrt

def distance(x1,y1,x2,y2):
    d12 = sqrt((x1-x2)**2 + (y1-y2)**2)
    return d12
    #Code below here with the same level of indentation will *never* run
    d12 = "Doesn't matter - I'll never run"
    print("Ill also never run")

print(distance(0, 0, 1, 1))
````

The arguments are identified by _order_, so the first argument will become x1, the second argument y1, and so on.  Change the order of what you provide when you call the function, and you'll get a different answer.



````python
def print_me(x,y):
    print('Hello')
    print('x is',x)
    print('y is',y)
    print('The End')

a = 1
b = 2
print_me(a,b)
print_me(b,a)
````

Things to notice:
 - We changed the order that we put the arguments in, and the values of x and y inside the function changed.
 - This function does not have a return value.  It has no `return` statement.  The function ends when the indent stops.

 However, note the code below -- we can override the ordering defined in the function header by explicitly labelling the input arguments. If we do this, the order doesn't matter.

````python
def print_me(x,y):
    print('Hello')
    print('x is',x)
    print('y is',y)
    print('The End')

a = 1
b = 2
print_me(x=a,y=b)
print_me(y=b,x=a)
````

I can also provide a "default" value for an argument when I define a function. This in turn makes the argument *optional*.

````python
def power(base, exponent=2):
    return base ** exponent

#This computers 3^2 as exponent defaults to 2.
print(power(3))
#This computers 3^3 as we have overwritten the exponent by providing a second input. 
print(power(3, 3))
#We can achieve the same by explicitly naming the arguments
print(power(exponent=3, base=3))
````

````python
#However, calling without any inputs will cause an error - as base has no default, we must provide a value for it.
print(power())
````

Here is an example with two inputs and two outputs.  Its goal is to sort two numbers into ascending order.

````python
def in_ascending_order(x,y):
    return min(x, y), max(x, y)

lo,hi = in_ascending_order(3,1)
print(lo,hi)
lo,hi = in_ascending_order(3,100)
print(lo,hi)
````

Things to notice:
- I can return two values by putting them comma separated, after `return`
- If two values come back, I store them in two different places

### Docstrings

When writing a function, it's good practice to include a docstring just below the first line. A docstring is a short description of what a function does. We expect every function you write in this course to include a docstring.

````python
def power(base, exponent=2):
    """Returns base raised to the power of exponent"""
    return base ** exponent
````

When you're reading code that's been written by someone else (or e.g. an AI model), reading the docstring will tell you what the function is *supposed* to do. Of course, the code may not actually do what the docstring does (check!), but they are often a good place to start when trying to understand what a function does or how it works.

## Example - Using functions to process vector quantities

Let's look at an example of a situation where we may want to use functions to organise our code. 

Let's consider a 2 component vector. We can represent this with two numbers -- say (x, y). As you know, there are many vector operations that we may want to perform -- addition, subtraction, dot and cross products and so on. Let's see how we can use functions to implement these operations.

We'll start by defining some example vectors we can use to test our code.

````python
v1_x = 1 # This is the 'x' component of vector 1
v1_y = 1 # This is the 'y' component of vector 1

v2_x = 1 # This is the 'x' component of vector 2
v2_y = 3 # This is the 'y' component of vector 2
````

Now let's try writing our first function for performing vector addition. Remembering that vector addition is performed component-wise - i.e $c = a + b = (a_x + b_x, a_y + b_y)$, we can write this by defining a function as below. As our function needs two vectors as input, and each vector consists of two values (the x- and y-components), our function needs four inputs. Similarly, we need to output two values -- the x-component and y-component of the vector given by a + b. So our function has two return values (c_x and c_y). 

````python
def vector_add(v1_x, v1_y, v2_x, v2_y):
    return v1_x + v2_x, v1_y + v2_y

v3_x, v3_y= vector_add(v1_x, v1_y, v2_x, v2_y)
print(v3_x, v3_y)
````

We may also want to calculate the dot product of two vectors. We can do this as below. This time we still need four inputs, but only return a single value.

````python
def dot_product(v1_x, v1_y, v2_x, v2_y):
    return (v1_x * v2_x) + (v1_y * v2_y)

print(dot_product(v1_x, v1_y, v2_x, v2_y))
````

## Scope

Mostly, the variables inside a function aren't the same as those outside.  Confused?  Believe me, this is a good thing, as it prevents accidental conflicts if you've used the same name for something elsewhere.  This is the concept of _scope_ of a variable - from where in the programme it can be accessed.

> Imagine the function as its own private little world.  It knows the values of its arguments, but nothing else.  It has its own little memory and can make its own variables, quite separate from the world outside that called it.

> This topic is more intricate in Python than there is time here to cover.  The interpretation above is safest for now - we'll show you a more complete picture in the seminar.

An illustrative example.  Watch what happens to variable `a`.

````python
def quadruple_me(x):
    print('Starting quadruple_me.  x is',x)
    a = 4*x
    print('Finishing quadruple_me.  a is',a)
    return a

a = 3
print('a is',a)
print('About to call quadruple_me')
b = quadruple_me(2)
print('Just called quadruple_me')
print('a is',a)
````

Eh? What? Why does `a` change back to 3?  There's no line that does that?

No there isn't - but there are _two_ variables called `a`.
* One is in the _global_ scope, and that one is set to 3 at the start, and is still set to 3 at the end.
* The other is in the _local_ scope of the function `quadruple_me`.  It is set to 4*2 when the function runs.  It lives in a different bit of memory than the `a` outside.


Another consequence of this is that variables defined in a function only exist while the function is being run. If the function terminates (because of a Return statement or reaching the end of the function body), any associated variables are removed from memory and trying to access them will lead to an error.

````python
A = "This is a string" 

def make_lower_case(txt):
    B = txt.lower()

make_lower_case(A)
print(B)
````

We get an error here because the variable B only exists while the function is running -- after line 4, the function terminates and B is removed from memory. Trying to use it later on line 7, leads to an error.
